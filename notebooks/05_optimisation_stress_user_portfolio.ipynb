```python
# =========================================================
# DYNAMIC INDIAN EQUITY DATA LOADER
# Supports arbitrary NSE/BSE portfolios
# =========================================================

from datetime import timedelta


def validate_portfolio_weights(portfolio_weights):
    """
    Validate and normalise a ticker-weight dictionary.
    """

    if not isinstance(portfolio_weights, dict):
        raise TypeError(
            "Portfolio must be supplied as a dictionary."
        )

    if len(portfolio_weights) < 2:
        raise ValueError(
            "Select at least two listed stocks."
        )

    cleaned_weights = {}

    for ticker, weight in portfolio_weights.items():

        clean_ticker = str(ticker).strip().upper()

        if not clean_ticker.endswith((".NS", ".BO")):
            raise ValueError(
                f"{clean_ticker}: use .NS for NSE or .BO for BSE."
            )

        numeric_weight = float(weight)

        if numeric_weight < 0:
            raise ValueError(
                f"{clean_ticker}: weights cannot be negative."
            )

        cleaned_weights[clean_ticker] = numeric_weight

    total_weight = sum(cleaned_weights.values())

    if total_weight <= 0:
        raise ValueError(
            "Portfolio weights must sum to more than zero."
        )

    normalised_weights = {
        ticker: weight / total_weight
        for ticker, weight in cleaned_weights.items()
    }

    return normalised_weights


def download_portfolio_market_data(
    portfolio_weights,
    start_date,
    end_date,
    benchmark=DEFAULT_BENCHMARK,
    minimum_observations=252,
):
    """
    Download adjusted market prices for an arbitrary Indian portfolio.

    The supplied end date is treated as inclusive.
    """

    validated_weights = validate_portfolio_weights(
        portfolio_weights
    )

    portfolio_tickers = list(
        validated_weights.keys()
    )

    download_tickers = (
        portfolio_tickers
        + [benchmark]
    )

    end_date_exclusive = (
        pd.Timestamp(end_date)
        + pd.Timedelta(days=1)
    ).strftime("%Y-%m-%d")

    raw_market_data = yf.download(
        tickers=download_tickers,
        start=start_date,
        end=end_date_exclusive,
        auto_adjust=True,
        progress=False,
        group_by="column",
        threads=True,
    )

    if raw_market_data.empty:
        raise RuntimeError(
            "No market data were downloaded."
        )

    if isinstance(
        raw_market_data.columns,
        pd.MultiIndex,
    ):

        if "Close" not in (
            raw_market_data.columns
            .get_level_values(0)
        ):
            raise RuntimeError(
                "Adjusted closing prices were not returned."
            )

        adjusted_close = (
            raw_market_data["Close"]
            .copy()
        )

    else:

        adjusted_close = (
            raw_market_data[["Close"]]
            .copy()
        )

        adjusted_close.columns = (
            download_tickers[:1]
        )

    adjusted_close = (
        adjusted_close
        .sort_index()
        .replace(
            [np.inf, -np.inf],
            np.nan,
        )
    )

    # Build a data-quality report before alignment.
    quality_records = []

    for ticker in download_tickers:

        if ticker in adjusted_close.columns:

            ticker_prices = (
                adjusted_close[ticker]
                .dropna()
            )

            observations = len(
                ticker_prices
            )

            first_date = (
                ticker_prices.index.min()
                if observations
                else pd.NaT
            )

            last_date = (
                ticker_prices.index.max()
                if observations
                else pd.NaT
            )

            missing_percentage = (
                adjusted_close[ticker]
                .isna()
                .mean()
            )

        else:

            observations = 0
            first_date = pd.NaT
            last_date = pd.NaT
            missing_percentage = 1.0

        quality_records.append(
            {
                "Ticker": ticker,
                "Role": (
                    "Benchmark"
                    if ticker == benchmark
                    else "Portfolio Holding"
                ),
                "Observations": observations,
                "First Date": first_date,
                "Last Date": last_date,
                "Missing Percentage":
                    missing_percentage,
                "Sufficient History":
                    observations
                    >= minimum_observations,
            }
        )

    data_quality = (
        pd.DataFrame(
            quality_records
        )
        .set_index("Ticker")
    )

    insufficient_tickers = (
        data_quality.index[
            ~data_quality[
                "Sufficient History"
            ]
        ]
        .tolist()
    )

    if insufficient_tickers:
        raise ValueError(
            "Insufficient price history for: "
            + ", ".join(insufficient_tickers)
        )

    missing_columns = [
        ticker
        for ticker in download_tickers
        if ticker not in adjusted_close.columns
    ]

    if missing_columns:
        raise ValueError(
            "No valid price series for: "
            + ", ".join(missing_columns)
        )

    # Fill only very short gaps, then retain common observations.
    aligned_prices = (
        adjusted_close[
            download_tickers
        ]
        .ffill(limit=3)
        .dropna()
    )

    daily_returns = (
        aligned_prices
        .pct_change(fill_method=None)
        .dropna()
    )

    portfolio_prices = (
        aligned_prices[
            portfolio_tickers
        ]
    )

    portfolio_returns = (
        daily_returns[
            portfolio_tickers
        ]
    )

    benchmark_prices = (
        aligned_prices[
            benchmark
        ]
    )

    benchmark_returns = (
        daily_returns[
            benchmark
        ]
    )

    validated_weight_series = (
        pd.Series(
            validated_weights,
            name="Portfolio Weight",
        )
        .reindex(portfolio_tickers)
    )

    return {
        "weights":
            validated_weight_series,

        "all_prices":
            aligned_prices,

        "portfolio_prices":
            portfolio_prices,

        "benchmark_prices":
            benchmark_prices,

        "all_returns":
            daily_returns,

        "portfolio_returns":
            portfolio_returns,

        "benchmark_returns":
            benchmark_returns,

        "data_quality":
            data_quality,
    }


# ---------------------------------------------------------
# Load the default India 10 research portfolio
# ---------------------------------------------------------

market_data = download_portfolio_market_data(
    portfolio_weights=DEFAULT_PORTFOLIO,
    start_date=DEFAULT_START_DATE,
    end_date=DEFAULT_END_DATE,
    benchmark=DEFAULT_BENCHMARK,
)

portfolio_weights = market_data["weights"]
portfolio_prices = market_data["portfolio_prices"]
portfolio_returns = market_data["portfolio_returns"]
benchmark_prices = market_data["benchmark_prices"]
benchmark_returns = market_data["benchmark_returns"]
data_quality = market_data["data_quality"]


# ---------------------------------------------------------
# Validation
# ---------------------------------------------------------

assert len(portfolio_weights) == 10
assert np.isclose(
    portfolio_weights.sum(),
    1.0,
)
assert not portfolio_returns.isna().any().any()
assert len(portfolio_returns) >= 252


print("Dynamic market-data loader validated.")
print("-" * 65)
print(
    "Portfolio holdings:",
    len(portfolio_weights),
)
print(
    "Aligned price observations:",
    len(portfolio_prices),
)
print(
    "Return observations:",
    len(portfolio_returns),
)
print(
    "Analysis period:",
    portfolio_prices.index.min().date(),
    "to",
    portfolio_prices.index.max().date(),
)
print(
    "Portfolio weights sum:",
    f"{portfolio_weights.sum():.2%}",
)

display(
    data_quality.style.format(
        {
            "Missing Percentage": "{:.2%}",
        }
    )
)
```

```python
# =========================================================
# BASELINE PORTFOLIO ANALYTICS AND OPTIMISATION INPUTS
# =========================================================

def calculate_maximum_drawdown(return_series):
    """
    Calculate the largest peak-to-trough portfolio loss.
    """

    wealth_index = (
        1.0 + return_series
    ).cumprod()

    running_peak = wealth_index.cummax()

    drawdown = (
        wealth_index
        / running_peak
        - 1.0
    )

    return float(drawdown.min())


def calculate_sortino_ratio(
    return_series,
    risk_free_rate=RISK_FREE_RATE,
):
    """
    Calculate annualised Sortino ratio using downside deviation.
    """

    daily_risk_free_rate = (
        (1.0 + risk_free_rate)
        ** (1.0 / TRADING_DAYS_PER_YEAR)
        - 1.0
    )

    excess_returns = (
        return_series
        - daily_risk_free_rate
    )

    downside_returns = excess_returns[
        excess_returns < 0
    ]

    if len(downside_returns) == 0:
        return np.nan

    downside_deviation = (
        downside_returns.std(ddof=1)
        * np.sqrt(TRADING_DAYS_PER_YEAR)
    )

    if downside_deviation == 0:
        return np.nan

    annualised_excess_return = (
        excess_returns.mean()
        * TRADING_DAYS_PER_YEAR
    )

    return float(
        annualised_excess_return
        / downside_deviation
    )


def calculate_portfolio_statistics(
    asset_returns,
    weights,
    benchmark_return_series=None,
    risk_free_rate=RISK_FREE_RATE,
):
    """
    Calculate performance and risk statistics for a portfolio.

    The calculation assumes the target weights are maintained through
    daily rebalancing. Periodic rebalancing will be added later.
    """

    weight_series = (
        pd.Series(
            weights,
            dtype=float,
        )
        .reindex(
            asset_returns.columns
        )
    )

    if weight_series.isna().any():
        raise ValueError(
            "Portfolio weights do not match the return columns."
        )

    if not np.isclose(
        weight_series.sum(),
        1.0,
    ):
        raise ValueError(
            "Portfolio weights must sum to 100%."
        )

    portfolio_daily_returns = (
        asset_returns
        .mul(
            weight_series,
            axis=1,
        )
        .sum(axis=1)
        .rename("Portfolio Return")
    )

    observations = len(
        portfolio_daily_returns
    )

    years = (
        observations
        / TRADING_DAYS_PER_YEAR
    )

    cumulative_growth = (
        1.0 + portfolio_daily_returns
    ).prod()

    total_return = (
        cumulative_growth
        - 1.0
    )

    cagr = (
        cumulative_growth
        ** (1.0 / years)
        - 1.0
    )

    annualised_volatility = (
        portfolio_daily_returns.std(ddof=1)
        * np.sqrt(TRADING_DAYS_PER_YEAR)
    )

    sharpe_ratio = (
        (cagr - risk_free_rate)
        / annualised_volatility
        if annualised_volatility > 0
        else np.nan
    )

    sortino_ratio = calculate_sortino_ratio(
        portfolio_daily_returns,
        risk_free_rate,
    )

    maximum_drawdown = calculate_maximum_drawdown(
        portfolio_daily_returns
    )

    calmar_ratio = (
        cagr
        / abs(maximum_drawdown)
        if maximum_drawdown < 0
        else np.nan
    )

    beta = np.nan
    correlation = np.nan

    if benchmark_return_series is not None:

        aligned_returns = pd.concat(
            [
                portfolio_daily_returns,
                benchmark_return_series.rename(
                    "Benchmark Return"
                ),
            ],
            axis=1,
        ).dropna()

        covariance_with_benchmark = (
            aligned_returns
            .cov()
            .loc[
                "Portfolio Return",
                "Benchmark Return",
            ]
        )

        benchmark_variance = (
            aligned_returns[
                "Benchmark Return"
            ]
            .var(ddof=1)
        )

        if benchmark_variance > 0:

            beta = (
                covariance_with_benchmark
                / benchmark_variance
            )

        correlation = (
            aligned_returns[
                "Portfolio Return"
            ]
            .corr(
                aligned_returns[
                    "Benchmark Return"
                ]
            )
        )

    ending_value = (
        INITIAL_INVESTMENT_INR
        * cumulative_growth
    )

    statistics = pd.Series(
        {
            "Observations":
                observations,

            "Ending Value (₹)":
                ending_value,

            "Total Return":
                total_return,

            "CAGR":
                cagr,

            "Annualised Volatility":
                annualised_volatility,

            "Sharpe Ratio":
                sharpe_ratio,

            "Sortino Ratio":
                sortino_ratio,

            "Maximum Drawdown":
                maximum_drawdown,

            "Calmar Ratio":
                calmar_ratio,

            "Beta vs Nifty 50":
                beta,

            "Correlation vs Nifty 50":
                correlation,
        },
        name="Current Portfolio",
    )

    return (
        portfolio_daily_returns,
        statistics,
    )


def calculate_diversification_statistics(
    weights,
    annualised_covariance,
):
    """
    Calculate portfolio concentration and diversification measures.
    """

    weight_series = pd.Series(
        weights,
        dtype=float,
    ).reindex(
        annualised_covariance.columns
    )

    portfolio_variance = float(
        weight_series.to_numpy().T
        @ annualised_covariance.to_numpy()
        @ weight_series.to_numpy()
    )

    portfolio_volatility = np.sqrt(
        portfolio_variance
    )

    individual_volatilities = np.sqrt(
        np.diag(
            annualised_covariance
        )
    )

    weighted_average_volatility = float(
        weight_series.to_numpy()
        @ individual_volatilities
    )

    diversification_ratio = (
        weighted_average_volatility
        / portfolio_volatility
        if portfolio_volatility > 0
        else np.nan
    )

    hhi = float(
        np.square(
            weight_series
        ).sum()
    )

    effective_number_of_holdings = (
        1.0 / hhi
        if hhi > 0
        else np.nan
    )

    correlation_matrix = (
        annualised_covariance
        .div(
            individual_volatilities,
            axis=0,
        )
        .div(
            individual_volatilities,
            axis=1,
        )
    )

    upper_triangle = (
        correlation_matrix
        .where(
            np.triu(
                np.ones(
                    correlation_matrix.shape
                ),
                k=1,
            ).astype(bool)
        )
        .stack()
    )

    return pd.Series(
        {
            "Number of Holdings":
                len(weight_series),

            "Largest Holding":
                weight_series.max(),

            "Top-3 Concentration":
                weight_series
                .nlargest(3)
                .sum(),

            "Herfindahl Index":
                hhi,

            "Effective Holdings":
                effective_number_of_holdings,

            "Average Pairwise Correlation":
                upper_triangle.mean(),

            "Diversification Ratio":
                diversification_ratio,
        },
        name="Current Portfolio",
    )


# ---------------------------------------------------------
# Estimate expected returns and covariance
# ---------------------------------------------------------

historical_expected_returns = (
    portfolio_returns.mean()
    * TRADING_DAYS_PER_YEAR
)

sample_annualised_covariance = (
    portfolio_returns.cov()
    * TRADING_DAYS_PER_YEAR
)

ledoit_wolf_estimator = LedoitWolf().fit(
    portfolio_returns.to_numpy()
)

ledoit_wolf_annualised_covariance = (
    pd.DataFrame(
        ledoit_wolf_estimator.covariance_,
        index=portfolio_returns.columns,
        columns=portfolio_returns.columns,
    )
    * TRADING_DAYS_PER_YEAR
)

asset_analytics = pd.DataFrame(
    {
        "Current Weight":
            portfolio_weights,

        "Expected Return":
            historical_expected_returns,

        "Annualised Volatility":
            portfolio_returns.std(ddof=1)
            * np.sqrt(
                TRADING_DAYS_PER_YEAR
            ),

        "Minimum Daily Return":
            portfolio_returns.min(),

        "Maximum Daily Return":
            portfolio_returns.max(),
    }
)

asset_analytics[
    "Return-to-Risk Ratio"
] = (
    asset_analytics[
        "Expected Return"
    ]
    / asset_analytics[
        "Annualised Volatility"
    ]
)


# ---------------------------------------------------------
# Evaluate the current portfolio
# ---------------------------------------------------------

(
    current_portfolio_returns,
    current_portfolio_statistics,
) = calculate_portfolio_statistics(
    asset_returns=portfolio_returns,
    weights=portfolio_weights,
    benchmark_return_series=benchmark_returns,
)

diversification_statistics = (
    calculate_diversification_statistics(
        weights=portfolio_weights,
        annualised_covariance=(
            ledoit_wolf_annualised_covariance
        ),
    )
)


# ---------------------------------------------------------
# Build cumulative wealth series
# ---------------------------------------------------------

current_portfolio_value = (
    INITIAL_INVESTMENT_INR
    * (
        1.0
        + current_portfolio_returns
    ).cumprod()
).rename(
    "Current Portfolio"
)

benchmark_value = (
    INITIAL_INVESTMENT_INR
    * (
        1.0
        + benchmark_returns
    ).cumprod()
).rename(
    "Nifty 50"
)

wealth_comparison = pd.concat(
    [
        current_portfolio_value,
        benchmark_value,
    ],
    axis=1,
).dropna()


# ---------------------------------------------------------
# Validation
# ---------------------------------------------------------

assert (
    ledoit_wolf_annualised_covariance.shape
    == (10, 10)
)

assert np.all(
    np.linalg.eigvalsh(
        ledoit_wolf_annualised_covariance
    ) >= -1e-10
)

assert not current_portfolio_returns.isna().any()


print("Baseline portfolio analytics completed.")
print("-" * 65)
print(
    "Expected-return method:",
    "Annualised historical arithmetic mean",
)
print(
    "Optimisation covariance:",
    "Ledoit-Wolf shrinkage covariance",
)
print(
    "Current portfolio CAGR:",
    f"{current_portfolio_statistics['CAGR']:.2%}",
)
print(
    "Current portfolio volatility:",
    f"{current_portfolio_statistics['Annualised Volatility']:.2%}",
)
print(
    "Current portfolio Sharpe:",
    f"{current_portfolio_statistics['Sharpe Ratio']:.2f}",
)
print(
    "Maximum drawdown:",
    f"{current_portfolio_statistics['Maximum Drawdown']:.2%}",
)
print(
    "Effective holdings:",
    f"{diversification_statistics['Effective Holdings']:.2f}",
)
print(
    "Diversification ratio:",
    f"{diversification_statistics['Diversification Ratio']:.2f}",
)

display(
    current_portfolio_statistics
    .to_frame("Result")
    .style.format(
        {
            "Result": lambda value: (
                f"₹{value:,.0f}"
                if value
                == current_portfolio_statistics[
                    "Ending Value (₹)"
                ]
                else f"{value:.4f}"
            )
        }
    )
)

display(
    diversification_statistics
    .to_frame("Result")
)

display(
    asset_analytics.style.format(
        {
            "Current Weight": "{:.2%}",
            "Expected Return": "{:.2%}",
            "Annualised Volatility": "{:.2%}",
            "Minimum Daily Return": "{:.2%}",
            "Maximum Daily Return": "{:.2%}",
            "Return-to-Risk Ratio": "{:.2f}",
        }
    )
)
```

```python
# =========================================================
# CORE PORTFOLIO OPTIMISATION ENGINE
# Minimum Volatility | Maximum Sharpe | Risk Parity
# =========================================================

# ---------------------------------------------------------
# Optimisation assumptions
# ---------------------------------------------------------

MINIMUM_STOCK_WEIGHT = 0.00
MAXIMUM_STOCK_WEIGHT = 0.20

# Historical returns are noisy, so shrink each estimate
# halfway toward the portfolio-wide average expected return.
EXPECTED_RETURN_SHRINKAGE = 0.50


# ---------------------------------------------------------
# Prepare robust optimisation inputs
# ---------------------------------------------------------

cross_sectional_mean_return = (
    historical_expected_returns.mean()
)

shrunk_expected_returns = (
    (
        1.0
        - EXPECTED_RETURN_SHRINKAGE
    )
    * historical_expected_returns
    + EXPECTED_RETURN_SHRINKAGE
    * cross_sectional_mean_return
)

optimisation_covariance = (
    ledoit_wolf_annualised_covariance.copy()
)

asset_names = (
    portfolio_returns.columns.tolist()
)

number_of_assets = len(
    asset_names
)

initial_weights = np.repeat(
    1.0 / number_of_assets,
    number_of_assets,
)

weight_bounds = [
    (
        MINIMUM_STOCK_WEIGHT,
        MAXIMUM_STOCK_WEIGHT,
    )
    for _ in range(number_of_assets)
]

full_investment_constraint = {
    "type": "eq",
    "fun": lambda weights: (
        np.sum(weights) - 1.0
    ),
}


# ---------------------------------------------------------
# Portfolio-model functions
# ---------------------------------------------------------

def model_portfolio_return(
    weights,
    expected_returns,
):
    """
    Calculate model-implied annual expected return.
    """

    return float(
        np.asarray(weights)
        @ expected_returns.to_numpy()
    )


def model_portfolio_volatility(
    weights,
    covariance_matrix,
):
    """
    Calculate model-implied annual portfolio volatility.
    """

    weight_array = np.asarray(
        weights,
        dtype=float,
    )

    portfolio_variance = float(
        weight_array.T
        @ covariance_matrix.to_numpy()
        @ weight_array
    )

    return float(
        np.sqrt(
            max(
                portfolio_variance,
                0.0,
            )
        )
    )


def negative_sharpe_objective(
    weights,
    expected_returns,
    covariance_matrix,
    risk_free_rate,
):
    """
    Negative Sharpe ratio for numerical minimisation.
    """

    expected_return = (
        model_portfolio_return(
            weights,
            expected_returns,
        )
    )

    volatility = (
        model_portfolio_volatility(
            weights,
            covariance_matrix,
        )
    )

    if volatility <= 0:
        return 1e10

    return -(
        (
            expected_return
            - risk_free_rate
        )
        / volatility
    )


def minimum_volatility_objective(
    weights,
    covariance_matrix,
):
    """
    Portfolio volatility objective.
    """

    return model_portfolio_volatility(
        weights,
        covariance_matrix,
    )


def calculate_percentage_risk_contributions(
    weights,
    covariance_matrix,
):
    """
    Calculate each holding's percentage contribution
    to total portfolio variance.
    """

    weight_array = np.asarray(
        weights,
        dtype=float,
    )

    covariance_array = (
        covariance_matrix.to_numpy()
    )

    marginal_variance = (
        covariance_array
        @ weight_array
    )

    total_variance = float(
        weight_array.T
        @ covariance_array
        @ weight_array
    )

    if total_variance <= 0:

        return np.repeat(
            np.nan,
            len(weight_array),
        )

    component_variance = (
        weight_array
        * marginal_variance
    )

    return (
        component_variance
        / total_variance
    )


def risk_parity_objective(
    weights,
    covariance_matrix,
):
    """
    Minimise differences between percentage risk contributions.
    """

    risk_contributions = (
        calculate_percentage_risk_contributions(
            weights,
            covariance_matrix,
        )
    )

    target_risk_contribution = (
        1.0
        / len(weights)
    )

    return float(
        np.sum(
            (
                risk_contributions
                - target_risk_contribution
            )
            ** 2
        )
    )


def solve_optimisation(
    objective_function,
    objective_arguments,
    portfolio_name,
):
    """
    Solve and validate one constrained portfolio optimisation.
    """

    result = minimize(
        fun=objective_function,
        x0=initial_weights,
        args=objective_arguments,
        method="SLSQP",
        bounds=weight_bounds,
        constraints=[
            full_investment_constraint
        ],
        options={
            "maxiter": 2000,
            "ftol": 1e-12,
            "disp": False,
        },
    )

    if not result.success:

        raise RuntimeError(
            f"{portfolio_name} optimisation failed: "
            f"{result.message}"
        )

    solved_weights = pd.Series(
        result.x,
        index=asset_names,
        name=portfolio_name,
    )

    # Remove tiny numerical values.
    solved_weights[
        solved_weights.abs() < 1e-8
    ] = 0.0

    solved_weights = (
        solved_weights
        / solved_weights.sum()
    )

    if not np.isclose(
        solved_weights.sum(),
        1.0,
        atol=1e-8,
    ):
        raise RuntimeError(
            f"{portfolio_name} weights do not sum to 100%."
        )

    if (
        solved_weights.max()
        > MAXIMUM_STOCK_WEIGHT + 1e-6
    ):
        raise RuntimeError(
            f"{portfolio_name} exceeds the maximum stock weight."
        )

    return solved_weights


# ---------------------------------------------------------
# Solve the three optimised portfolios
# ---------------------------------------------------------

minimum_volatility_weights = (
    solve_optimisation(
        objective_function=(
            minimum_volatility_objective
        ),
        objective_arguments=(
            optimisation_covariance,
        ),
        portfolio_name=(
            "Minimum Volatility"
        ),
    )
)

maximum_sharpe_weights = (
    solve_optimisation(
        objective_function=(
            negative_sharpe_objective
        ),
        objective_arguments=(
            shrunk_expected_returns,
            optimisation_covariance,
            RISK_FREE_RATE,
        ),
        portfolio_name=(
            "Maximum Sharpe"
        ),
    )
)

risk_parity_weights = (
    solve_optimisation(
        objective_function=(
            risk_parity_objective
        ),
        objective_arguments=(
            optimisation_covariance,
        ),
        portfolio_name=(
            "Risk Parity"
        ),
    )
)


# ---------------------------------------------------------
# Consolidate portfolio weights
# ---------------------------------------------------------

optimised_portfolio_weights = pd.concat(
    [
        portfolio_weights.rename(
            "Current Equal Weight"
        ),
        minimum_volatility_weights,
        maximum_sharpe_weights,
        risk_parity_weights,
    ],
    axis=1,
)

optimised_portfolio_weights[
    "Maximum-Sharpe Risk Contribution"
] = (
    calculate_percentage_risk_contributions(
        maximum_sharpe_weights,
        optimisation_covariance,
    )
)

optimised_portfolio_weights[
    "Risk-Parity Risk Contribution"
] = (
    calculate_percentage_risk_contributions(
        risk_parity_weights,
        optimisation_covariance,
    )
)


# ---------------------------------------------------------
# Calculate realised historical performance
# ---------------------------------------------------------

portfolio_weight_sets = {
    "Current Equal Weight":
        portfolio_weights,

    "Minimum Volatility":
        minimum_volatility_weights,

    "Maximum Sharpe":
        maximum_sharpe_weights,

    "Risk Parity":
        risk_parity_weights,
}

historical_performance_records = {}
diversification_records = {}
model_estimate_records = {}

optimised_daily_returns = {}

for portfolio_name, weights in (
    portfolio_weight_sets.items()
):

    (
        portfolio_return_series,
        portfolio_statistics,
    ) = calculate_portfolio_statistics(
        asset_returns=portfolio_returns,
        weights=weights,
        benchmark_return_series=(
            benchmark_returns
        ),
    )

    optimised_daily_returns[
        portfolio_name
    ] = portfolio_return_series

    historical_performance_records[
        portfolio_name
    ] = portfolio_statistics

    diversification_records[
        portfolio_name
    ] = (
        calculate_diversification_statistics(
            weights=weights,
            annualised_covariance=(
                optimisation_covariance
            ),
        )
    )

    model_expected_return = (
        model_portfolio_return(
            weights,
            shrunk_expected_returns,
        )
    )

    model_volatility = (
        model_portfolio_volatility(
            weights,
            optimisation_covariance,
        )
    )

    model_sharpe = (
        (
            model_expected_return
            - RISK_FREE_RATE
        )
        / model_volatility
        if model_volatility > 0
        else np.nan
    )

    model_estimate_records[
        portfolio_name
    ] = {
        "Model Expected Return":
            model_expected_return,

        "Model Volatility":
            model_volatility,

        "Model Sharpe":
            model_sharpe,
    }


historical_performance_comparison = (
    pd.DataFrame(
        historical_performance_records
    )
    .T
)

diversification_comparison = (
    pd.DataFrame(
        diversification_records
    )
    .T
)

model_estimate_comparison = (
    pd.DataFrame(
        model_estimate_records
    )
    .T
)


# ---------------------------------------------------------
# Validation
# ---------------------------------------------------------

for portfolio_name in (
    portfolio_weight_sets
):

    assert np.isclose(
        optimised_portfolio_weights[
            portfolio_name
        ].sum(),
        1.0,
    )

assert (
    minimum_volatility_weights.max()
    <= MAXIMUM_STOCK_WEIGHT + 1e-6
)

assert (
    maximum_sharpe_weights.max()
    <= MAXIMUM_STOCK_WEIGHT + 1e-6
)

assert (
    risk_parity_weights.max()
    <= MAXIMUM_STOCK_WEIGHT + 1e-6
)


# ---------------------------------------------------------
# Results
# ---------------------------------------------------------

print("Core portfolio optimisation completed.")
print("-" * 70)
print(
    "Constraint:",
    (
        f"Long-only, maximum "
        f"{MAXIMUM_STOCK_WEIGHT:.0%} per stock"
    ),
)
print(
    "Expected-return shrinkage:",
    f"{EXPECTED_RETURN_SHRINKAGE:.0%}",
)
print(
    "Covariance estimator:",
    "Ledoit-Wolf shrinkage",
)
print(
    "Minimum-volatility portfolio volatility:",
    (
        f"{model_estimate_comparison.loc['Minimum Volatility', 'Model Volatility']:.2%}"
    ),
)
print(
    "Maximum-Sharpe model Sharpe:",
    (
        f"{model_estimate_comparison.loc['Maximum Sharpe', 'Model Sharpe']:.2f}"
    ),
)
print(
    "Largest risk-parity weight:",
    f"{risk_parity_weights.max():.2%}",
)

print("\nOPTIMISED WEIGHTS")

display(
    optimised_portfolio_weights.style.format(
        {
            "Current Equal Weight": "{:.2%}",
            "Minimum Volatility": "{:.2%}",
            "Maximum Sharpe": "{:.2%}",
            "Risk Parity": "{:.2%}",
            "Maximum-Sharpe Risk Contribution": "{:.2%}",
            "Risk-Parity Risk Contribution": "{:.2%}",
        }
    )
)

print("\nMODEL-IMPLIED ESTIMATES")

display(
    model_estimate_comparison.style.format(
        {
            "Model Expected Return": "{:.2%}",
            "Model Volatility": "{:.2%}",
            "Model Sharpe": "{:.2f}",
        }
    )
)

print("\nREALISED HISTORICAL PERFORMANCE")

display(
    historical_performance_comparison[
        [
            "Ending Value (₹)",
            "CAGR",
            "Annualised Volatility",
            "Sharpe Ratio",
            "Maximum Drawdown",
            "Beta vs Nifty 50",
        ]
    ].style.format(
        {
            "Ending Value (₹)": "₹{:,.0f}",
            "CAGR": "{:.2%}",
            "Annualised Volatility": "{:.2%}",
            "Sharpe Ratio": "{:.2f}",
            "Maximum Drawdown": "{:.2%}",
            "Beta vs Nifty 50": "{:.2f}",
        }
    )
)

print(
    "\nNote: these are in-sample historical comparisons. "
    "Out-of-sample and rolling optimisation tests will follow."
)
```

```python
# =========================================================
# CONSTRAINED EFFICIENT FRONTIER
# Long-only | Fully invested | Maximum 20% per stock
# =========================================================

# ---------------------------------------------------------
# 1. Find the highest feasible expected-return portfolio
# ---------------------------------------------------------

maximum_return_result = minimize(
    fun=lambda weights: -model_portfolio_return(
        weights,
        shrunk_expected_returns,
    ),
    x0=maximum_sharpe_weights.to_numpy(),
    method="SLSQP",
    bounds=weight_bounds,
    constraints=[
        full_investment_constraint
    ],
    options={
        "maxiter": 3000,
        "ftol": 1e-12,
        "disp": False,
    },
)

if not maximum_return_result.success:
    raise RuntimeError(
        "Maximum-return optimisation failed: "
        + maximum_return_result.message
    )

maximum_return_weights = pd.Series(
    maximum_return_result.x,
    index=asset_names,
    name="Maximum Feasible Return",
)

maximum_return_weights[
    maximum_return_weights.abs() < 1e-8
] = 0.0

maximum_return_weights = (
    maximum_return_weights
    / maximum_return_weights.sum()
)


# ---------------------------------------------------------
# 2. Define the feasible efficient-frontier range
# ---------------------------------------------------------

minimum_frontier_return = (
    model_portfolio_return(
        minimum_volatility_weights,
        shrunk_expected_returns,
    )
)

maximum_frontier_return = (
    model_portfolio_return(
        maximum_return_weights,
        shrunk_expected_returns,
    )
)

target_returns = np.linspace(
    minimum_frontier_return,
    maximum_frontier_return - 1e-6,
    60,
)


# ---------------------------------------------------------
# 3. Solve minimum volatility for each target return
# ---------------------------------------------------------

frontier_records = []
frontier_weight_records = []

starting_weights = (
    minimum_volatility_weights
    .to_numpy()
)

for target_return in target_returns:

    target_return_constraint = {
        "type": "eq",
        "fun": lambda weights, target=target_return: (
            model_portfolio_return(
                weights,
                shrunk_expected_returns,
            )
            - target
        ),
    }

    frontier_result = minimize(
        fun=minimum_volatility_objective,
        x0=starting_weights,
        args=(
            optimisation_covariance,
        ),
        method="SLSQP",
        bounds=weight_bounds,
        constraints=[
            full_investment_constraint,
            target_return_constraint,
        ],
        options={
            "maxiter": 3000,
            "ftol": 1e-12,
            "disp": False,
        },
    )

    if not frontier_result.success:
        continue

    frontier_weights = pd.Series(
        frontier_result.x,
        index=asset_names,
    )

    frontier_weights[
        frontier_weights.abs() < 1e-8
    ] = 0.0

    frontier_weights = (
        frontier_weights
        / frontier_weights.sum()
    )

    expected_return = (
        model_portfolio_return(
            frontier_weights,
            shrunk_expected_returns,
        )
    )

    volatility = (
        model_portfolio_volatility(
            frontier_weights,
            optimisation_covariance,
        )
    )

    sharpe_ratio = (
        (
            expected_return
            - RISK_FREE_RATE
        )
        / volatility
        if volatility > 0
        else np.nan
    )

    frontier_number = len(
        frontier_records
    ) + 1

    frontier_records.append(
        {
            "Frontier Portfolio":
                frontier_number,

            "Target Return":
                target_return,

            "Expected Return":
                expected_return,

            "Volatility":
                volatility,

            "Sharpe Ratio":
                sharpe_ratio,

            "Largest Weight":
                frontier_weights.max(),

            "Active Holdings":
                int(
                    (
                        frontier_weights
                        > 0.001
                    ).sum()
                ),
        }
    )

    weight_record = {
        "Frontier Portfolio":
            frontier_number,
    }

    weight_record.update(
        frontier_weights.to_dict()
    )

    frontier_weight_records.append(
        weight_record
    )

    starting_weights = (
        frontier_weights.to_numpy()
    )


efficient_frontier = (
    pd.DataFrame(
        frontier_records
    )
    .set_index(
        "Frontier Portfolio"
    )
)

efficient_frontier_weights = (
    pd.DataFrame(
        frontier_weight_records
    )
    .set_index(
        "Frontier Portfolio"
    )
)


# ---------------------------------------------------------
# 4. Identify the best Sharpe point on the frontier
# ---------------------------------------------------------

best_frontier_index = (
    efficient_frontier[
        "Sharpe Ratio"
    ]
    .idxmax()
)

best_frontier_portfolio = (
    efficient_frontier.loc[
        best_frontier_index
    ]
)

best_frontier_weights = (
    efficient_frontier_weights.loc[
        best_frontier_index
    ]
)


# ---------------------------------------------------------
# 5. Build comparison points for the chart
# ---------------------------------------------------------

comparison_point_records = []

for portfolio_name, weights in (
    portfolio_weight_sets.items()
):

    point_return = (
        model_portfolio_return(
            weights,
            shrunk_expected_returns,
        )
    )

    point_volatility = (
        model_portfolio_volatility(
            weights,
            optimisation_covariance,
        )
    )

    point_sharpe = (
        (
            point_return
            - RISK_FREE_RATE
        )
        / point_volatility
    )

    comparison_point_records.append(
        {
            "Portfolio":
                portfolio_name,

            "Expected Return":
                point_return,

            "Volatility":
                point_volatility,

            "Sharpe Ratio":
                point_sharpe,
        }
    )

comparison_points = (
    pd.DataFrame(
        comparison_point_records
    )
    .set_index("Portfolio")
)


# ---------------------------------------------------------
# 6. Plot the efficient frontier
# ---------------------------------------------------------

efficient_frontier_figure = go.Figure()

efficient_frontier_figure.add_trace(
    go.Scatter(
        x=(
            efficient_frontier[
                "Volatility"
            ]
            * 100
        ),
        y=(
            efficient_frontier[
                "Expected Return"
            ]
            * 100
        ),
        mode="lines+markers",
        name="Efficient Frontier",
        text=[
            (
                f"Sharpe: {sharpe:.2f}"
                f"<br>Active holdings: {active}"
            )
            for sharpe, active in zip(
                efficient_frontier[
                    "Sharpe Ratio"
                ],
                efficient_frontier[
                    "Active Holdings"
                ],
            )
        ],
        hovertemplate=(
            "Volatility: %{x:.2f}%"
            "<br>Expected return: %{y:.2f}%"
            "<br>%{text}"
            "<extra></extra>"
        ),
    )
)

for portfolio_name, row in (
    comparison_points.iterrows()
):

    efficient_frontier_figure.add_trace(
        go.Scatter(
            x=[
                row["Volatility"]
                * 100
            ],
            y=[
                row["Expected Return"]
                * 100
            ],
            mode="markers+text",
            name=portfolio_name,
            text=[
                portfolio_name
            ],
            textposition="top center",
            hovertemplate=(
                f"{portfolio_name}"
                "<br>Volatility: %{x:.2f}%"
                "<br>Expected return: %{y:.2f}%"
                f"<br>Sharpe: {row['Sharpe Ratio']:.2f}"
                "<extra></extra>"
            ),
        )
    )

efficient_frontier_figure.update_layout(
    title=(
        "India 10 Constrained Efficient Frontier"
    ),
    xaxis_title=(
        "Model Annualised Volatility (%)"
    ),
    yaxis_title=(
        "Model Expected Return (%)"
    ),
    height=650,
    hovermode="closest",
)

efficient_frontier_figure.show()


# ---------------------------------------------------------
# 7. Validation
# ---------------------------------------------------------

assert len(
    efficient_frontier
) >= 50

assert np.allclose(
    efficient_frontier_weights.sum(axis=1),
    1.0,
    atol=1e-6,
)

assert (
    efficient_frontier_weights.max().max()
    <= MAXIMUM_STOCK_WEIGHT + 1e-5
)

assert (
    best_frontier_portfolio[
        "Sharpe Ratio"
    ]
    > 0
)


# ---------------------------------------------------------
# 8. Results
# ---------------------------------------------------------

print("Constrained efficient frontier completed.")
print("-" * 70)
print(
    "Successful frontier portfolios:",
    len(efficient_frontier),
)
print(
    "Feasible expected-return range:",
    (
        f"{minimum_frontier_return:.2%}"
        f" to "
        f"{maximum_frontier_return:.2%}"
    ),
)
print(
    "Minimum frontier volatility:",
    (
        f"{efficient_frontier['Volatility'].min():.2%}"
    ),
)
print(
    "Best frontier Sharpe:",
    (
        f"{best_frontier_portfolio['Sharpe Ratio']:.2f}"
    ),
)
print(
    "Best frontier expected return:",
    (
        f"{best_frontier_portfolio['Expected Return']:.2%}"
    ),
)
print(
    "Best frontier volatility:",
    (
        f"{best_frontier_portfolio['Volatility']:.2%}"
    ),
)
print(
    "Best frontier active holdings:",
    int(
        best_frontier_portfolio[
            "Active Holdings"
        ]
    ),
)

print("\nBEST FRONTIER WEIGHTS")

display(
    best_frontier_weights
    .sort_values(
        ascending=False
    )
    .to_frame(
        "Weight"
    )
    .style.format(
        {
            "Weight": "{:.2%}",
        }
    )
)

print("\nPORTFOLIO COMPARISON POINTS")

display(
    comparison_points.style.format(
        {
            "Expected Return": "{:.2%}",
            "Volatility": "{:.2%}",
            "Sharpe Ratio": "{:.2f}",
        }
    )
)
```

```python
# =========================================================
# ROLLING OUT-OF-SAMPLE PORTFOLIO OPTIMISATION
# Monthly rebalancing | No look-ahead | Transaction costs
# =========================================================

OUT_OF_SAMPLE_START_DATE = "2024-01-01"
ROLLING_TRAINING_WINDOW = 756
MINIMUM_TRAINING_OBSERVATIONS = 504
TRANSACTION_COST_RATE = 0.0015


# ---------------------------------------------------------
# 1. Generic rolling optimisation functions
# ---------------------------------------------------------

def estimate_rolling_inputs(
    training_returns,
):
    """
    Estimate shrunk expected returns and Ledoit-Wolf covariance
    using information available before one rebalance date.
    """

    historical_returns = (
        training_returns.mean()
        * TRADING_DAYS_PER_YEAR
    )

    cross_sectional_mean = (
        historical_returns.mean()
    )

    expected_returns = (
        (
            1.0
            - EXPECTED_RETURN_SHRINKAGE
        )
        * historical_returns
        + EXPECTED_RETURN_SHRINKAGE
        * cross_sectional_mean
    )

    covariance_estimator = (
        LedoitWolf()
        .fit(
            training_returns.to_numpy()
        )
    )

    covariance_matrix = (
        pd.DataFrame(
            covariance_estimator.covariance_,
            index=training_returns.columns,
            columns=training_returns.columns,
        )
        * TRADING_DAYS_PER_YEAR
    )

    return (
        expected_returns,
        covariance_matrix,
    )


def solve_rolling_portfolio(
    portfolio_type,
    expected_returns,
    covariance_matrix,
    starting_weights,
):
    """
    Solve one rolling portfolio using only historical information.
    """

    number_of_assets = len(
        expected_returns
    )

    bounds = [
        (
            MINIMUM_STOCK_WEIGHT,
            MAXIMUM_STOCK_WEIGHT,
        )
        for _ in range(number_of_assets)
    ]

    investment_constraint = {
        "type": "eq",
        "fun": lambda weights: (
            np.sum(weights) - 1.0
        ),
    }

    if portfolio_type == "Minimum Volatility":

        objective_function = (
            minimum_volatility_objective
        )

        objective_arguments = (
            covariance_matrix,
        )

    elif portfolio_type == "Maximum Sharpe":

        objective_function = (
            negative_sharpe_objective
        )

        objective_arguments = (
            expected_returns,
            covariance_matrix,
            RISK_FREE_RATE,
        )

    elif portfolio_type == "Risk Parity":

        objective_function = (
            risk_parity_objective
        )

        objective_arguments = (
            covariance_matrix,
        )

    else:

        raise ValueError(
            f"Unknown portfolio type: {portfolio_type}"
        )

    optimisation_result = minimize(
        fun=objective_function,
        x0=np.asarray(
            starting_weights,
            dtype=float,
        ),
        args=objective_arguments,
        method="SLSQP",
        bounds=bounds,
        constraints=[
            investment_constraint
        ],
        options={
            "maxiter": 3000,
            "ftol": 1e-12,
            "disp": False,
        },
    )

    if not optimisation_result.success:

        raise RuntimeError(
            f"{portfolio_type} failed: "
            f"{optimisation_result.message}"
        )

    solved_weights = pd.Series(
        optimisation_result.x,
        index=expected_returns.index,
        name=portfolio_type,
    )

    solved_weights[
        solved_weights.abs() < 1e-8
    ] = 0.0

    solved_weights = (
        solved_weights
        / solved_weights.sum()
    )

    return solved_weights


# ---------------------------------------------------------
# 2. Create monthly execution dates
# ---------------------------------------------------------

out_of_sample_returns = (
    portfolio_returns
    .loc[
        OUT_OF_SAMPLE_START_DATE:
    ]
    .copy()
)

monthly_execution_dates = (
    out_of_sample_returns
    .groupby(
        out_of_sample_returns.index.to_period(
            "M"
        )
    )
    .head(1)
    .index
)

if len(monthly_execution_dates) == 0:

    raise RuntimeError(
        "No out-of-sample rebalance dates were found."
    )


# ---------------------------------------------------------
# 3. Generate point-in-time target weights
# ---------------------------------------------------------

rolling_portfolio_names = [
    "Current Equal Weight",
    "Minimum Volatility",
    "Maximum Sharpe",
    "Risk Parity",
]

previous_solution = {
    "Minimum Volatility":
        portfolio_weights.copy(),

    "Maximum Sharpe":
        portfolio_weights.copy(),

    "Risk Parity":
        portfolio_weights.copy(),
}

rolling_target_records = []
rolling_estimation_records = []

for execution_date in monthly_execution_dates:

    historical_window = (
        portfolio_returns
        .loc[
            portfolio_returns.index
            < execution_date
        ]
        .tail(
            ROLLING_TRAINING_WINDOW
        )
    )

    if (
        len(historical_window)
        < MINIMUM_TRAINING_OBSERVATIONS
    ):
        continue

    signal_date = (
        historical_window.index[-1]
    )

    (
        rolling_expected_returns,
        rolling_covariance,
    ) = estimate_rolling_inputs(
        historical_window
    )

    monthly_weight_sets = {
        "Current Equal Weight":
            portfolio_weights.copy(),
    }

    for portfolio_name in [
        "Minimum Volatility",
        "Maximum Sharpe",
        "Risk Parity",
    ]:

        solved_weights = (
            solve_rolling_portfolio(
                portfolio_type=portfolio_name,
                expected_returns=(
                    rolling_expected_returns
                ),
                covariance_matrix=(
                    rolling_covariance
                ),
                starting_weights=(
                    previous_solution[
                        portfolio_name
                    ]
                ),
            )
        )

        monthly_weight_sets[
            portfolio_name
        ] = solved_weights

        previous_solution[
            portfolio_name
        ] = solved_weights

    for portfolio_name, weights in (
        monthly_weight_sets.items()
    ):

        for ticker, weight in (
            weights.items()
        ):

            rolling_target_records.append(
                {
                    "Execution Date":
                        execution_date,

                    "Signal Date":
                        signal_date,

                    "Portfolio":
                        portfolio_name,

                    "Ticker":
                        ticker,

                    "Target Weight":
                        weight,

                    "Training Observations":
                        len(
                            historical_window
                        ),
                }
            )

    rolling_estimation_records.append(
        {
            "Execution Date":
                execution_date,

            "Signal Date":
                signal_date,

            "Training Observations":
                len(
                    historical_window
                ),

            "Minimum Expected Return":
                rolling_expected_returns.min(),

            "Maximum Expected Return":
                rolling_expected_returns.max(),

            "Average Expected Return":
                rolling_expected_returns.mean(),
        }
    )


rolling_target_weights = pd.DataFrame(
    rolling_target_records
)

rolling_estimation_summary = pd.DataFrame(
    rolling_estimation_records
)

valid_execution_dates = (
    rolling_target_weights[
        "Execution Date"
    ]
    .drop_duplicates()
    .sort_values()
    .tolist()
)

if len(valid_execution_dates) == 0:

    raise RuntimeError(
        "No valid rolling portfolios were generated."
    )


# ---------------------------------------------------------
# 4. Realistic monthly-rebalanced backtest
# ---------------------------------------------------------

def backtest_monthly_targets(
    asset_returns,
    target_weight_data,
    portfolio_name,
    transaction_cost_rate,
):
    """
    Backtest monthly target weights with drift between rebalances.

    Transaction cost equals the cost rate multiplied by one-way
    turnover. Initial portfolio establishment is not charged.
    """

    portfolio_targets = (
        target_weight_data.loc[
            target_weight_data[
                "Portfolio"
            ]
            == portfolio_name
        ]
        .pivot(
            index="Execution Date",
            columns="Ticker",
            values="Target Weight",
        )
        .reindex(
            columns=asset_returns.columns
        )
        .sort_index()
    )

    first_execution_date = (
        portfolio_targets.index.min()
    )

    test_returns = (
        asset_returns
        .loc[
            first_execution_date:
        ]
        .copy()
    )

    current_weights = None
    portfolio_value = 1.0

    daily_records = []
    trading_records = []

    for return_date, daily_asset_returns in (
        test_returns.iterrows()
    ):

        starting_value = (
            portfolio_value
        )

        transaction_cost = 0.0
        one_way_turnover = 0.0
        rebalanced = False

        if return_date in portfolio_targets.index:

            target_weights = (
                portfolio_targets
                .loc[
                    return_date
                ]
                .astype(float)
            )

            if current_weights is None:

                current_weights = (
                    target_weights.copy()
                )

            else:

                one_way_turnover = float(
                    0.5
                    * np.abs(
                        target_weights
                        - current_weights
                    ).sum()
                )

                transaction_cost = (
                    transaction_cost_rate
                    * one_way_turnover
                )

                portfolio_value *= (
                    1.0
                    - transaction_cost
                )

                current_weights = (
                    target_weights.copy()
                )

            rebalanced = True

            trading_records.append(
                {
                    "Date":
                        return_date,

                    "Portfolio":
                        portfolio_name,

                    "One-Way Turnover":
                        one_way_turnover,

                    "Transaction Cost Rate":
                        transaction_cost,

                    "Transaction Cost (₹)":
                        (
                            starting_value
                            * INITIAL_INVESTMENT_INR
                            * transaction_cost
                        ),
                }
            )

        gross_daily_return = float(
            current_weights
            .mul(
                daily_asset_returns
            )
            .sum()
        )

        portfolio_value *= (
            1.0
            + gross_daily_return
        )

        net_daily_return = (
            portfolio_value
            / starting_value
            - 1.0
        )

        post_return_values = (
            current_weights
            * (
                1.0
                + daily_asset_returns
            )
        )

        current_weights = (
            post_return_values
            / post_return_values.sum()
        )

        daily_records.append(
            {
                "Date":
                    return_date,

                "Portfolio":
                    portfolio_name,

                "Gross Return":
                    gross_daily_return,

                "Net Return":
                    net_daily_return,

                "Portfolio Value (₹)":
                    (
                        INITIAL_INVESTMENT_INR
                        * portfolio_value
                    ),

                "Rebalanced":
                    rebalanced,

                "One-Way Turnover":
                    one_way_turnover,

                "Transaction Cost Rate":
                    transaction_cost,
            }
        )

    return (
        pd.DataFrame(
            daily_records
        ).set_index("Date"),
        pd.DataFrame(
            trading_records
        ),
    )


rolling_daily_results = []
rolling_trading_results = []

for portfolio_name in rolling_portfolio_names:

    (
        strategy_daily,
        strategy_trading,
    ) = backtest_monthly_targets(
        asset_returns=portfolio_returns,
        target_weight_data=(
            rolling_target_weights
        ),
        portfolio_name=portfolio_name,
        transaction_cost_rate=(
            TRANSACTION_COST_RATE
        ),
    )

    rolling_daily_results.append(
        strategy_daily.reset_index()
    )

    rolling_trading_results.append(
        strategy_trading
    )


rolling_backtest_daily = pd.concat(
    rolling_daily_results,
    ignore_index=True,
)

rolling_trading_history = pd.concat(
    rolling_trading_results,
    ignore_index=True,
)


# ---------------------------------------------------------
# 5. Performance statistics for realised return series
# ---------------------------------------------------------

def calculate_return_series_statistics(
    return_series,
    benchmark_series,
):
    """
    Calculate realised statistics for one net return series.
    """

    aligned_data = pd.concat(
        [
            return_series.rename(
                "Portfolio"
            ),
            benchmark_series.rename(
                "Benchmark"
            ),
        ],
        axis=1,
    ).dropna()

    portfolio_series = (
        aligned_data["Portfolio"]
    )

    benchmark_aligned = (
        aligned_data["Benchmark"]
    )

    observations = len(
        portfolio_series
    )

    years = (
        observations
        / TRADING_DAYS_PER_YEAR
    )

    cumulative_growth = (
        1.0
        + portfolio_series
    ).prod()

    total_return = (
        cumulative_growth
        - 1.0
    )

    cagr = (
        cumulative_growth
        ** (1.0 / years)
        - 1.0
    )

    volatility = (
        portfolio_series.std(ddof=1)
        * np.sqrt(
            TRADING_DAYS_PER_YEAR
        )
    )

    sharpe = (
        (
            cagr
            - RISK_FREE_RATE
        )
        / volatility
        if volatility > 0
        else np.nan
    )

    maximum_drawdown = (
        calculate_maximum_drawdown(
            portfolio_series
        )
    )

    covariance = (
        aligned_data.cov().loc[
            "Portfolio",
            "Benchmark",
        ]
    )

    benchmark_variance = (
        benchmark_aligned.var(ddof=1)
    )

    beta = (
        covariance
        / benchmark_variance
        if benchmark_variance > 0
        else np.nan
    )

    return pd.Series(
        {
            "Observations":
                observations,

            "Ending Value (₹)":
                (
                    INITIAL_INVESTMENT_INR
                    * cumulative_growth
                ),

            "Total Return":
                total_return,

            "CAGR":
                cagr,

            "Annualised Volatility":
                volatility,

            "Sharpe Ratio":
                sharpe,

            "Sortino Ratio":
                calculate_sortino_ratio(
                    portfolio_series
                ),

            "Maximum Drawdown":
                maximum_drawdown,

            "Calmar Ratio":
                (
                    cagr
                    / abs(
                        maximum_drawdown
                    )
                    if maximum_drawdown < 0
                    else np.nan
                ),

            "Beta vs Nifty 50":
                beta,

            "Correlation vs Nifty 50":
                portfolio_series.corr(
                    benchmark_aligned
                ),
        }
    )


first_backtest_date = min(
    valid_execution_dates
)

benchmark_oos_returns = (
    benchmark_returns
    .loc[
        first_backtest_date:
    ]
)

rolling_performance_records = {}

for portfolio_name in rolling_portfolio_names:

    strategy_returns = (
        rolling_backtest_daily.loc[
            rolling_backtest_daily[
                "Portfolio"
            ]
            == portfolio_name
        ]
        .set_index("Date")[
            "Net Return"
        ]
    )

    rolling_performance_records[
        portfolio_name
    ] = (
        calculate_return_series_statistics(
            return_series=(
                strategy_returns
            ),
            benchmark_series=(
                benchmark_oos_returns
            ),
        )
    )


nifty_statistics = (
    calculate_return_series_statistics(
        return_series=(
            benchmark_oos_returns
        ),
        benchmark_series=(
            benchmark_oos_returns
        ),
    )
)

rolling_performance_records[
    "Nifty 50"
] = nifty_statistics

rolling_performance_comparison = (
    pd.DataFrame(
        rolling_performance_records
    )
    .T
)


# ---------------------------------------------------------
# 6. Trading summary
# ---------------------------------------------------------

rolling_trading_summary = (
    rolling_trading_history
    .groupby("Portfolio")
    .agg(
        Rebalances=(
            "Date",
            "count",
        ),

        Total_One_Way_Turnover=(
            "One-Way Turnover",
            "sum",
        ),

        Average_Turnover=(
            "One-Way Turnover",
            "mean",
        ),

        Total_Transaction_Costs_INR=(
            "Transaction Cost (₹)",
            "sum",
        ),
    )
)

backtest_years = (
    len(
        benchmark_oos_returns
    )
    / TRADING_DAYS_PER_YEAR
)

rolling_trading_summary[
    "Annualised One-Way Turnover"
] = (
    rolling_trading_summary[
        "Total_One_Way_Turnover"
    ]
    / backtest_years
)

rolling_trading_summary[
    "Costs as % of Initial Capital"
] = (
    rolling_trading_summary[
        "Total_Transaction_Costs_INR"
    ]
    / INITIAL_INVESTMENT_INR
)


# ---------------------------------------------------------
# 7. Validation
# ---------------------------------------------------------

assert (
    rolling_target_weights[
        "Execution Date"
    ].nunique()
    >= 24
)

assert np.allclose(
    rolling_target_weights
    .groupby(
        [
            "Execution Date",
            "Portfolio",
        ]
    )[
        "Target Weight"
    ]
    .sum(),
    1.0,
    atol=1e-6,
)

assert (
    rolling_target_weights[
        "Target Weight"
    ].max()
    <= MAXIMUM_STOCK_WEIGHT + 1e-6
)

assert not (
    rolling_backtest_daily[
        "Net Return"
    ]
    .isna()
    .any()
)


# ---------------------------------------------------------
# 8. Results
# ---------------------------------------------------------

print(
    "Rolling out-of-sample optimisation completed."
)
print("-" * 72)
print(
    "First execution date:",
    pd.Timestamp(
        first_backtest_date
    ).date(),
)
print(
    "Latest observation:",
    rolling_backtest_daily[
        "Date"
    ].max().date(),
)
print(
    "Monthly rebalances:",
    rolling_target_weights[
        "Execution Date"
    ].nunique(),
)
print(
    "Training window:",
    (
        f"up to {ROLLING_TRAINING_WINDOW} "
        "trading observations"
    ),
)
print(
    "Minimum training history:",
    MINIMUM_TRAINING_OBSERVATIONS,
)
print(
    "Transaction cost:",
    (
        f"{TRANSACTION_COST_RATE:.2%} "
        "per one-way turnover"
    ),
)
print(
    "Look-ahead bias:",
    "None — every signal uses data before execution",
)

print("\nOUT-OF-SAMPLE PERFORMANCE")

display(
    rolling_performance_comparison[
        [
            "Ending Value (₹)",
            "CAGR",
            "Annualised Volatility",
            "Sharpe Ratio",
            "Maximum Drawdown",
            "Beta vs Nifty 50",
        ]
    ].style.format(
        {
            "Ending Value (₹)": "₹{:,.0f}",
            "CAGR": "{:.2%}",
            "Annualised Volatility": "{:.2%}",
            "Sharpe Ratio": "{:.2f}",
            "Maximum Drawdown": "{:.2%}",
            "Beta vs Nifty 50": "{:.2f}",
        }
    )
)

print("\nTURNOVER AND TRANSACTION COSTS")

display(
    rolling_trading_summary.style.format(
        {
            "Total_One_Way_Turnover":
                "{:.2%}",

            "Average_Turnover":
                "{:.2%}",

            "Total_Transaction_Costs_INR":
                "₹{:,.2f}",

            "Annualised One-Way Turnover":
                "{:.2%}",

            "Costs as % of Initial Capital":
                "{:.2%}",
        }
    )
)
```

```python
# =========================================================
# HISTORICAL STRESS TESTING
# Data-driven stress windows | Out-of-sample | After costs
# =========================================================

# ---------------------------------------------------------
# 1. Create aligned realised return matrix
# ---------------------------------------------------------

rolling_backtest_daily[
    "Date"
] = pd.to_datetime(
    rolling_backtest_daily["Date"]
)

strategy_return_matrix = (
    rolling_backtest_daily
    .pivot(
        index="Date",
        columns="Portfolio",
        values="Net Return",
    )
    .sort_index()
)

stress_return_matrix = pd.concat(
    [
        strategy_return_matrix,
        benchmark_oos_returns.rename(
            "Nifty 50"
        ),
    ],
    axis=1,
).dropna()

stress_portfolios = (
    stress_return_matrix.columns.tolist()
)

nifty_stress_returns = (
    stress_return_matrix["Nifty 50"]
)


# ---------------------------------------------------------
# 2. Helper functions
# ---------------------------------------------------------

def calculate_window_return(
    return_series,
):
    """
    Calculate compounded return over one period.
    """

    return float(
        (
            1.0
            + return_series
        ).prod()
        - 1.0
    )


def calculate_rolling_compounded_return(
    return_series,
    window,
):
    """
    Calculate rolling compounded returns.
    """

    return (
        (
            1.0
            + return_series
        )
        .rolling(window)
        .apply(
            np.prod,
            raw=True,
        )
        - 1.0
    )


def get_window_start_date(
    index,
    end_date,
    window,
):
    """
    Locate the first date of a rolling window.
    """

    end_position = index.get_loc(
        end_date
    )

    start_position = max(
        0,
        end_position
        - window
        + 1,
    )

    return index[
        start_position
    ]


def identify_drawdown_episode(
    return_series,
):
    """
    Identify peak, trough and recovery dates for maximum drawdown.
    """

    wealth_index = (
        1.0
        + return_series
    ).cumprod()

    running_peak = (
        wealth_index.cummax()
    )

    drawdown_series = (
        wealth_index
        / running_peak
        - 1.0
    )

    trough_date = (
        drawdown_series.idxmin()
    )

    peak_date = (
        wealth_index
        .loc[:trough_date]
        .idxmax()
    )

    peak_value = float(
        wealth_index.loc[
            peak_date
        ]
    )

    post_trough_wealth = (
        wealth_index.loc[
            trough_date:
        ]
    )

    recovery_candidates = (
        post_trough_wealth[
            post_trough_wealth
            >= peak_value
        ]
    )

    recovery_date = (
        recovery_candidates.index[0]
        if len(
            recovery_candidates
        ) > 0
        else pd.NaT
    )

    peak_to_trough_days = (
        return_series
        .loc[
            peak_date:
            trough_date
        ]
        .shape[0]
        - 1
    )

    recovery_days = (
        return_series
        .loc[
            trough_date:
            recovery_date
        ]
        .shape[0]
        - 1
        if pd.notna(
            recovery_date
        )
        else np.nan
    )

    return {
        "Peak Date":
            peak_date,

        "Trough Date":
            trough_date,

        "Recovery Date":
            recovery_date,

        "Maximum Drawdown":
            float(
                drawdown_series.min()
            ),

        "Peak-to-Trough Trading Days":
            peak_to_trough_days,

        "Recovery Trading Days":
            recovery_days,
    }


# ---------------------------------------------------------
# 3. Identify data-driven Nifty stress periods
# ---------------------------------------------------------

nifty_21_day_returns = (
    calculate_rolling_compounded_return(
        nifty_stress_returns,
        window=21,
    )
)

nifty_63_day_returns = (
    calculate_rolling_compounded_return(
        nifty_stress_returns,
        window=63,
    )
)

nifty_63_day_volatility = (
    nifty_stress_returns
    .rolling(63)
    .std(ddof=1)
    * np.sqrt(
        TRADING_DAYS_PER_YEAR
    )
)

worst_single_day = (
    nifty_stress_returns.idxmin()
)

worst_21_day_end = (
    nifty_21_day_returns.idxmin()
)

worst_63_day_end = (
    nifty_63_day_returns.idxmin()
)

highest_volatility_end = (
    nifty_63_day_volatility.idxmax()
)

nifty_drawdown_episode = (
    identify_drawdown_episode(
        nifty_stress_returns
    )
)

stress_period_records = [
    {
        "Stress Scenario":
            "Worst Nifty Single Day",

        "Start Date":
            worst_single_day,

        "End Date":
            worst_single_day,

        "Selection Basis":
            "Lowest daily Nifty 50 return",
    },

    {
        "Stress Scenario":
            "Worst Nifty 21-Day Window",

        "Start Date":
            get_window_start_date(
                stress_return_matrix.index,
                worst_21_day_end,
                21,
            ),

        "End Date":
            worst_21_day_end,

        "Selection Basis":
            "Lowest compounded 21-day Nifty return",
    },

    {
        "Stress Scenario":
            "Worst Nifty 63-Day Window",

        "Start Date":
            get_window_start_date(
                stress_return_matrix.index,
                worst_63_day_end,
                63,
            ),

        "End Date":
            worst_63_day_end,

        "Selection Basis":
            "Lowest compounded 63-day Nifty return",
    },

    {
        "Stress Scenario":
            "Highest Nifty 63-Day Volatility",

        "Start Date":
            get_window_start_date(
                stress_return_matrix.index,
                highest_volatility_end,
                63,
            ),

        "End Date":
            highest_volatility_end,

        "Selection Basis":
            "Highest annualised 63-day Nifty volatility",
    },

    {
        "Stress Scenario":
            "Maximum Nifty Drawdown",

        "Start Date":
            nifty_drawdown_episode[
                "Peak Date"
            ],

        "End Date":
            nifty_drawdown_episode[
                "Trough Date"
            ],

        "Selection Basis":
            "Largest Nifty peak-to-trough decline",
    },
]

historical_stress_periods = (
    pd.DataFrame(
        stress_period_records
    )
)

historical_stress_periods[
    "Start Date"
] = pd.to_datetime(
    historical_stress_periods[
        "Start Date"
    ]
)

historical_stress_periods[
    "End Date"
] = pd.to_datetime(
    historical_stress_periods[
        "End Date"
    ]
)


# ---------------------------------------------------------
# 4. Evaluate every portfolio during each stress period
# ---------------------------------------------------------

stress_test_records = []

for _, scenario in (
    historical_stress_periods.iterrows()
):

    scenario_name = (
        scenario[
            "Stress Scenario"
        ]
    )

    start_date = (
        scenario[
            "Start Date"
        ]
    )

    end_date = (
        scenario[
            "End Date"
        ]
    )

    scenario_returns = (
        stress_return_matrix
        .loc[
            start_date:
            end_date
        ]
    )

    nifty_scenario_return = (
        calculate_window_return(
            scenario_returns[
                "Nifty 50"
            ]
        )
    )

    for portfolio_name in (
        stress_portfolios
    ):

        portfolio_scenario_returns = (
            scenario_returns[
                portfolio_name
            ]
        )

        cumulative_return = (
            calculate_window_return(
                portfolio_scenario_returns
            )
        )

        annualised_volatility = (
            portfolio_scenario_returns
            .std(ddof=1)
            * np.sqrt(
                TRADING_DAYS_PER_YEAR
            )
            if len(
                portfolio_scenario_returns
            ) > 1
            else np.nan
        )

        maximum_drawdown = (
            calculate_maximum_drawdown(
                portfolio_scenario_returns
            )
        )

        loss_reduction_vs_nifty = (
            cumulative_return
            - nifty_scenario_return
        )

        downside_capture = (
            cumulative_return
            / nifty_scenario_return
            if nifty_scenario_return < 0
            else np.nan
        )

        stress_test_records.append(
            {
                "Stress Scenario":
                    scenario_name,

                "Portfolio":
                    portfolio_name,

                "Start Date":
                    start_date,

                "End Date":
                    end_date,

                "Trading Days":
                    len(
                        portfolio_scenario_returns
                    ),

                "Cumulative Return":
                    cumulative_return,

                "Annualised Volatility":
                    annualised_volatility,

                "Maximum Drawdown":
                    maximum_drawdown,

                "Worst Daily Return":
                    portfolio_scenario_returns.min(),

                "Loss Reduction vs Nifty":
                    loss_reduction_vs_nifty,

                "Downside Capture":
                    downside_capture,
            }
        )


historical_stress_results = pd.DataFrame(
    stress_test_records
)


# ---------------------------------------------------------
# 5. Build portfolio drawdown-episode table
# ---------------------------------------------------------

drawdown_episode_records = []

for portfolio_name in (
    stress_portfolios
):

    episode = identify_drawdown_episode(
        stress_return_matrix[
            portfolio_name
        ]
    )

    drawdown_episode_records.append(
        {
            "Portfolio":
                portfolio_name,

            **episode,
        }
    )

portfolio_drawdown_episodes = (
    pd.DataFrame(
        drawdown_episode_records
    )
    .set_index("Portfolio")
)


# ---------------------------------------------------------
# 6. Create compact stress-return comparison
# ---------------------------------------------------------

stress_return_comparison = (
    historical_stress_results
    .pivot(
        index="Stress Scenario",
        columns="Portfolio",
        values="Cumulative Return",
    )
)

stress_drawdown_comparison = (
    historical_stress_results
    .pivot(
        index="Stress Scenario",
        columns="Portfolio",
        values="Maximum Drawdown",
    )
)


# ---------------------------------------------------------
# 7. Determine stress-test winners
# ---------------------------------------------------------

stress_winner_records = []

for scenario_name, scenario_data in (
    historical_stress_results
    .groupby(
        "Stress Scenario"
    )
):

    non_benchmark_data = (
        scenario_data.loc[
            scenario_data[
                "Portfolio"
            ]
            != "Nifty 50"
        ]
    )

    best_portfolio_row = (
        non_benchmark_data
        .sort_values(
            "Cumulative Return",
            ascending=False,
        )
        .iloc[0]
    )

    stress_winner_records.append(
        {
            "Stress Scenario":
                scenario_name,

            "Best Portfolio":
                best_portfolio_row[
                    "Portfolio"
                ],

            "Best Portfolio Return":
                best_portfolio_row[
                    "Cumulative Return"
                ],

            "Nifty 50 Return":
                scenario_data.loc[
                    scenario_data[
                        "Portfolio"
                    ]
                    == "Nifty 50",
                    "Cumulative Return",
                ].iloc[0],

            "Loss Reduction vs Nifty":
                best_portfolio_row[
                    "Loss Reduction vs Nifty"
                ],
        }
    )

stress_test_winners = (
    pd.DataFrame(
        stress_winner_records
    )
    .set_index(
        "Stress Scenario"
    )
)


# ---------------------------------------------------------
# 8. Validation
# ---------------------------------------------------------

assert len(
    historical_stress_periods
) == 5

assert (
    historical_stress_results[
        "Stress Scenario"
    ].nunique()
    == 5
)

assert (
    historical_stress_results[
        "Portfolio"
    ].nunique()
    == 5
)

assert not (
    historical_stress_results[
        "Cumulative Return"
    ]
    .isna()
    .any()
)


# ---------------------------------------------------------
# 9. Results
# ---------------------------------------------------------

print("Historical stress testing completed.")
print("-" * 72)
print(
    "Test period:",
    stress_return_matrix.index.min().date(),
    "to",
    stress_return_matrix.index.max().date(),
)
print(
    "Stress scenarios:",
    len(
        historical_stress_periods
    ),
)
print(
    "Portfolios tested:",
    len(
        stress_portfolios
    ),
)
print(
    "Method:",
    "Realised out-of-sample net returns after transaction costs",
)

print("\nIDENTIFIED STRESS PERIODS")

display(
    historical_stress_periods
)

print("\nSTRESS-PERIOD CUMULATIVE RETURNS")

display(
    stress_return_comparison.style.format(
        "{:.2%}"
    )
)

print("\nSTRESS-TEST WINNERS")

display(
    stress_test_winners.style.format(
        {
            "Best Portfolio Return":
                "{:.2%}",

            "Nifty 50 Return":
                "{:.2%}",

            "Loss Reduction vs Nifty":
                "{:+.2%}",
        }
    )
)

print("\nMAXIMUM DRAWDOWN EPISODES")

display(
    portfolio_drawdown_episodes.style.format(
        {
            "Maximum Drawdown":
                "{:.2%}",

            "Peak-to-Trough Trading Days":
                "{:.0f}",

            "Recovery Trading Days":
                "{:.0f}",
        },
        na_rep="Not recovered",
    )
)
```

```python
# =========================================================
# MONTE CARLO RISK SIMULATION
# Block bootstrap | VaR | Expected Shortfall | Loss probability
# =========================================================

MONTE_CARLO_SIMULATIONS = 10_000
MONTE_CARLO_HORIZON_DAYS = 252
BOOTSTRAP_BLOCK_LENGTH = 21
MONTE_CARLO_RANDOM_SEED = 42
SIMULATION_CHUNK_SIZE = 500


# ---------------------------------------------------------
# 1. Extract the latest point-in-time portfolio weights
# ---------------------------------------------------------

latest_weight_date = pd.to_datetime(
    rolling_target_weights[
        "Execution Date"
    ].max()
)

latest_portfolio_weights = (
    rolling_target_weights.loc[
        pd.to_datetime(
            rolling_target_weights[
                "Execution Date"
            ]
        )
        == latest_weight_date
    ]
    .pivot(
        index="Ticker",
        columns="Portfolio",
        values="Target Weight",
    )
    .reindex(
        index=portfolio_returns.columns,
        columns=rolling_portfolio_names,
    )
)

if latest_portfolio_weights.isna().any().any():
    raise RuntimeError(
        "Latest portfolio weights contain missing values."
    )

assert np.allclose(
    latest_portfolio_weights.sum(axis=0),
    1.0,
    atol=1e-6,
)


# ---------------------------------------------------------
# 2. Prepare the latest historical estimation window
# ---------------------------------------------------------

simulation_estimation_returns = (
    portfolio_returns
    .loc[
        portfolio_returns.index
        <= portfolio_returns.index.max()
    ]
    .tail(
        ROLLING_TRAINING_WINDOW
    )
    .copy()
)

if (
    len(simulation_estimation_returns)
    < MINIMUM_TRAINING_OBSERVATIONS
):
    raise RuntimeError(
        "Insufficient observations for Monte Carlo simulation."
    )

(
    simulation_expected_returns,
    simulation_covariance,
) = estimate_rolling_inputs(
    simulation_estimation_returns
)

# Adjust the historical daily returns so their means match the
# shrunk annual expected-return estimates.
target_daily_expected_returns = (
    simulation_expected_returns
    / TRADING_DAYS_PER_YEAR
)

adjusted_bootstrap_returns = (
    simulation_estimation_returns
    - simulation_estimation_returns.mean()
    + target_daily_expected_returns
)

adjusted_return_array = (
    adjusted_bootstrap_returns
    .to_numpy(
        dtype=float
    )
)

weight_matrix = (
    latest_portfolio_weights
    .to_numpy(
        dtype=float
    )
)

portfolio_names = (
    latest_portfolio_weights.columns.tolist()
)

number_of_portfolios = len(
    portfolio_names
)

number_of_assets = (
    adjusted_return_array.shape[1]
)

number_of_blocks = int(
    np.ceil(
        MONTE_CARLO_HORIZON_DAYS
        / BOOTSTRAP_BLOCK_LENGTH
    )
)

maximum_block_start = (
    len(adjusted_return_array)
    - BOOTSTRAP_BLOCK_LENGTH
)

if maximum_block_start < 1:
    raise RuntimeError(
        "Estimation history is too short for block bootstrap."
    )


# ---------------------------------------------------------
# 3. Run the simulation in memory-efficient chunks
# ---------------------------------------------------------

random_generator = np.random.default_rng(
    MONTE_CARLO_RANDOM_SEED
)

terminal_return_chunks = []
maximum_drawdown_chunks = []

simulations_completed = 0

while (
    simulations_completed
    < MONTE_CARLO_SIMULATIONS
):

    current_chunk_size = min(
        SIMULATION_CHUNK_SIZE,
        MONTE_CARLO_SIMULATIONS
        - simulations_completed,
    )

    block_start_indices = (
        random_generator.integers(
            low=0,
            high=maximum_block_start + 1,
            size=(
                current_chunk_size,
                number_of_blocks,
            ),
        )
    )

    block_offsets = np.arange(
        BOOTSTRAP_BLOCK_LENGTH
    )

    sampled_indices = (
        block_start_indices[
            :,
            :,
            None,
        ]
        + block_offsets[
            None,
            None,
            :,
        ]
    )

    sampled_indices = (
        sampled_indices
        .reshape(
            current_chunk_size,
            -1,
        )
        [
            :,
            :MONTE_CARLO_HORIZON_DAYS,
        ]
    )

    simulated_asset_returns = (
        adjusted_return_array[
            sampled_indices
        ]
    )

    # Buy-and-hold asset wealth paths.
    simulated_asset_wealth = np.cumprod(
        1.0
        + simulated_asset_returns,
        axis=1,
    )

    simulated_portfolio_wealth = np.einsum(
        "sda,ap->sdp",
        simulated_asset_wealth,
        weight_matrix,
    )

    simulated_terminal_returns = (
        simulated_portfolio_wealth[
            :,
            -1,
            :,
        ]
        - 1.0
    )

    simulated_running_peaks = np.maximum.accumulate(
        simulated_portfolio_wealth,
        axis=1,
    )

    simulated_drawdowns = (
        simulated_portfolio_wealth
        / simulated_running_peaks
        - 1.0
    )

    simulated_maximum_drawdowns = (
        simulated_drawdowns.min(
            axis=1
        )
    )

    terminal_return_chunks.append(
        simulated_terminal_returns
    )

    maximum_drawdown_chunks.append(
        simulated_maximum_drawdowns
    )

    simulations_completed += (
        current_chunk_size
    )


monte_carlo_terminal_returns = np.vstack(
    terminal_return_chunks
)

monte_carlo_maximum_drawdowns = np.vstack(
    maximum_drawdown_chunks
)


# ---------------------------------------------------------
# 4. VaR and Expected Shortfall helpers
# ---------------------------------------------------------

def calculate_var_and_expected_shortfall(
    return_array,
    confidence_level,
):
    """
    Return positive loss values for VaR and Expected Shortfall.
    """

    return_array = np.asarray(
        return_array,
        dtype=float,
    )

    tail_probability = (
        1.0
        - confidence_level
    )

    return_quantile = np.quantile(
        return_array,
        tail_probability,
    )

    tail_returns = return_array[
        return_array
        <= return_quantile
    ]

    value_at_risk = max(
        0.0,
        -float(
            return_quantile
        ),
    )

    expected_shortfall = max(
        0.0,
        -float(
            tail_returns.mean()
        ),
    )

    return (
        value_at_risk,
        expected_shortfall,
    )


# ---------------------------------------------------------
# 5. Calculate one-day historical risk
# ---------------------------------------------------------

latest_daily_portfolio_returns = (
    adjusted_bootstrap_returns
    @ latest_portfolio_weights
)

one_day_risk_records = []

for portfolio_name in portfolio_names:

    portfolio_daily_returns = (
        latest_daily_portfolio_returns[
            portfolio_name
        ]
        .to_numpy()
    )

    (
        var_95,
        es_95,
    ) = calculate_var_and_expected_shortfall(
        portfolio_daily_returns,
        confidence_level=0.95,
    )

    (
        var_99,
        es_99,
    ) = calculate_var_and_expected_shortfall(
        portfolio_daily_returns,
        confidence_level=0.99,
    )

    one_day_risk_records.append(
        {
            "Portfolio":
                portfolio_name,

            "1-Day VaR 95%":
                var_95,

            "1-Day ES 95%":
                es_95,

            "1-Day VaR 99%":
                var_99,

            "1-Day ES 99%":
                es_99,

            "Worst Historical Day":
                portfolio_daily_returns.min(),
        }
    )


one_day_risk_summary = (
    pd.DataFrame(
        one_day_risk_records
    )
    .set_index("Portfolio")
)


# ---------------------------------------------------------
# 6. Calculate one-year simulation risk
# ---------------------------------------------------------

monte_carlo_summary_records = []

for portfolio_number, portfolio_name in enumerate(
    portfolio_names
):

    terminal_returns = (
        monte_carlo_terminal_returns[
            :,
            portfolio_number,
        ]
    )

    maximum_drawdowns = (
        monte_carlo_maximum_drawdowns[
            :,
            portfolio_number,
        ]
    )

    (
        one_year_var_95,
        one_year_es_95,
    ) = calculate_var_and_expected_shortfall(
        terminal_returns,
        confidence_level=0.95,
    )

    (
        one_year_var_99,
        one_year_es_99,
    ) = calculate_var_and_expected_shortfall(
        terminal_returns,
        confidence_level=0.99,
    )

    monte_carlo_summary_records.append(
        {
            "Portfolio":
                portfolio_name,

            "Mean Terminal Return":
                terminal_returns.mean(),

            "Median Terminal Return":
                np.median(
                    terminal_returns
                ),

            "5th Percentile Return":
                np.quantile(
                    terminal_returns,
                    0.05,
                ),

            "95th Percentile Return":
                np.quantile(
                    terminal_returns,
                    0.95,
                ),

            "Probability of Loss":
                np.mean(
                    terminal_returns < 0
                ),

            "Probability of Loss > 10%":
                np.mean(
                    terminal_returns < -0.10
                ),

            "Probability of Loss > 20%":
                np.mean(
                    terminal_returns < -0.20
                ),

            "One-Year VaR 95%":
                one_year_var_95,

            "One-Year ES 95%":
                one_year_es_95,

            "One-Year VaR 99%":
                one_year_var_99,

            "One-Year ES 99%":
                one_year_es_99,

            "Median Maximum Drawdown":
                np.median(
                    maximum_drawdowns
                ),

            "5th Percentile Maximum Drawdown":
                np.quantile(
                    maximum_drawdowns,
                    0.05,
                ),

            "Median Terminal Value (₹)":
                (
                    INITIAL_INVESTMENT_INR
                    * (
                        1.0
                        + np.median(
                            terminal_returns
                        )
                    )
                ),

            "5th Percentile Terminal Value (₹)":
                (
                    INITIAL_INVESTMENT_INR
                    * (
                        1.0
                        + np.quantile(
                            terminal_returns,
                            0.05,
                        )
                    )
                ),
        }
    )


monte_carlo_risk_summary = (
    pd.DataFrame(
        monte_carlo_summary_records
    )
    .set_index("Portfolio")
)


# ---------------------------------------------------------
# 7. Build terminal-value distributions for export
# ---------------------------------------------------------

monte_carlo_terminal_value_records = []

for portfolio_number, portfolio_name in enumerate(
    portfolio_names
):

    portfolio_terminal_returns = (
        monte_carlo_terminal_returns[
            :,
            portfolio_number,
        ]
    )

    portfolio_drawdowns = (
        monte_carlo_maximum_drawdowns[
            :,
            portfolio_number,
        ]
    )

    for simulation_number in range(
        MONTE_CARLO_SIMULATIONS
    ):

        monte_carlo_terminal_value_records.append(
            {
                "Simulation":
                    simulation_number + 1,

                "Portfolio":
                    portfolio_name,

                "Terminal Return":
                    portfolio_terminal_returns[
                        simulation_number
                    ],

                "Terminal Value (₹)":
                    (
                        INITIAL_INVESTMENT_INR
                        * (
                            1.0
                            + portfolio_terminal_returns[
                                simulation_number
                            ]
                        )
                    ),

                "Maximum Drawdown":
                    portfolio_drawdowns[
                        simulation_number
                    ],
            }
        )


monte_carlo_terminal_values = pd.DataFrame(
    monte_carlo_terminal_value_records
)


# ---------------------------------------------------------
# 8. Plot terminal return distributions
# ---------------------------------------------------------

monte_carlo_distribution_figure = px.histogram(
    monte_carlo_terminal_values,
    x="Terminal Return",
    color="Portfolio",
    nbins=100,
    barmode="overlay",
    opacity=0.45,
    title=(
        "One-Year Monte Carlo Terminal-Return Distribution"
    ),
)

monte_carlo_distribution_figure.update_layout(
    xaxis_tickformat=".0%",
    xaxis_title="One-Year Terminal Return",
    yaxis_title="Simulation Frequency",
    height=650,
)

monte_carlo_distribution_figure.show()


# ---------------------------------------------------------
# 9. Validation
# ---------------------------------------------------------

assert (
    monte_carlo_terminal_returns.shape
    == (
        MONTE_CARLO_SIMULATIONS,
        number_of_portfolios,
    )
)

assert (
    monte_carlo_maximum_drawdowns.shape
    == (
        MONTE_CARLO_SIMULATIONS,
        number_of_portfolios,
    )
)

assert not np.isnan(
    monte_carlo_terminal_returns
).any()

assert np.all(
    monte_carlo_maximum_drawdowns
    <= 1e-10
)

assert (
    monte_carlo_terminal_values[
        "Simulation"
    ].nunique()
    == MONTE_CARLO_SIMULATIONS
)


# ---------------------------------------------------------
# 10. Results
# ---------------------------------------------------------

print("Monte Carlo risk simulation completed.")
print("-" * 72)
print(
    "Risk-estimation date:",
    simulation_estimation_returns.index.max().date(),
)
print(
    "Portfolio-weight date:",
    latest_weight_date.date(),
)
print(
    "Simulations:",
    f"{MONTE_CARLO_SIMULATIONS:,}",
)
print(
    "Forecast horizon:",
    (
        f"{MONTE_CARLO_HORIZON_DAYS} "
        "trading days"
    ),
)
print(
    "Bootstrap block length:",
    (
        f"{BOOTSTRAP_BLOCK_LENGTH} "
        "trading days"
    ),
)
print(
    "Estimation observations:",
    len(
        simulation_estimation_returns
    ),
)
print(
    "Method:",
    (
        "Cross-sectional historical block bootstrap "
        "with shrunk expected returns"
    ),
)
print(
    "Portfolio assumption:",
    "Buy-and-hold over the one-year simulation horizon",
)

print("\nONE-DAY HISTORICAL VaR AND EXPECTED SHORTFALL")

display(
    one_day_risk_summary.style.format(
        {
            "1-Day VaR 95%": "{:.2%}",
            "1-Day ES 95%": "{:.2%}",
            "1-Day VaR 99%": "{:.2%}",
            "1-Day ES 99%": "{:.2%}",
            "Worst Historical Day": "{:.2%}",
        }
    )
)

print("\nONE-YEAR MONTE CARLO RISK")

display(
    monte_carlo_risk_summary[
        [
            "Median Terminal Return",
            "5th Percentile Return",
            "95th Percentile Return",
            "Probability of Loss",
            "Probability of Loss > 10%",
            "One-Year VaR 95%",
            "One-Year ES 95%",
            "Median Maximum Drawdown",
            "5th Percentile Maximum Drawdown",
            "Median Terminal Value (₹)",
        ]
    ].style.format(
        {
            "Median Terminal Return": "{:.2%}",
            "5th Percentile Return": "{:.2%}",
            "95th Percentile Return": "{:.2%}",
            "Probability of Loss": "{:.2%}",
            "Probability of Loss > 10%": "{:.2%}",
            "One-Year VaR 95%": "{:.2%}",
            "One-Year ES 95%": "{:.2%}",
            "Median Maximum Drawdown": "{:.2%}",
            "5th Percentile Maximum Drawdown": "{:.2%}",
            "Median Terminal Value (₹)": "₹{:,.0f}",
        }
    )
)

print(
    "\nImportant: Monte Carlo results are model-based risk estimates, "
    "not forecasts or guarantees of future returns."
)
```

```python
# =========================================================
# BULL, BASE AND BEAR SCENARIO ANALYSIS
# Market-beta shocks | Stressed volatility | INR outcomes
# =========================================================

# ---------------------------------------------------------
# 1. Scenario assumptions
# ---------------------------------------------------------

SCENARIO_ASSUMPTIONS = {
    "Bear": {
        "Nifty 50 Return": -0.20,
        "Volatility Multiplier": 1.50,
    },

    "Base": {
        "Nifty 50 Return": 0.08,
        "Volatility Multiplier": 1.00,
    },

    "Bull": {
        "Nifty 50 Return": 0.20,
        "Volatility Multiplier": 0.85,
    },
}

SCENARIO_BASE_MARKET_RETURN = (
    SCENARIO_ASSUMPTIONS[
        "Base"
    ][
        "Nifty 50 Return"
    ]
)

SCENARIO_INITIAL_VALUE_INR = (
    INITIAL_INVESTMENT_INR
)


# ---------------------------------------------------------
# 2. Align latest estimation returns with Nifty 50
# ---------------------------------------------------------

scenario_estimation_data = pd.concat(
    [
        simulation_estimation_returns,
        benchmark_returns.rename(
            "Nifty 50"
        ),
    ],
    axis=1,
).dropna()

scenario_asset_returns = (
    scenario_estimation_data[
        portfolio_returns.columns
    ]
)

scenario_benchmark_returns = (
    scenario_estimation_data[
        "Nifty 50"
    ]
)

benchmark_variance = (
    scenario_benchmark_returns.var(
        ddof=1
    )
)

if benchmark_variance <= 0:
    raise RuntimeError(
        "Benchmark variance is invalid."
    )


# ---------------------------------------------------------
# 3. Estimate stock betas using latest available history
# ---------------------------------------------------------

stock_beta_records = []

for ticker in (
    scenario_asset_returns.columns
):

    covariance_with_market = (
        scenario_asset_returns[
            ticker
        ]
        .cov(
            scenario_benchmark_returns
        )
    )

    stock_beta = (
        covariance_with_market
        / benchmark_variance
    )

    stock_correlation = (
        scenario_asset_returns[
            ticker
        ]
        .corr(
            scenario_benchmark_returns
        )
    )

    stock_beta_records.append(
        {
            "Ticker":
                ticker,

            "Beta vs Nifty 50":
                stock_beta,

            "Correlation vs Nifty 50":
                stock_correlation,

            "Shrunk Expected Return":
                simulation_expected_returns[
                    ticker
                ],
        }
    )


scenario_stock_statistics = (
    pd.DataFrame(
        stock_beta_records
    )
    .set_index("Ticker")
)


# ---------------------------------------------------------
# 4. Calculate asset returns under each scenario
# ---------------------------------------------------------

scenario_asset_return_records = []

for scenario_name, assumptions in (
    SCENARIO_ASSUMPTIONS.items()
):

    market_return = (
        assumptions[
            "Nifty 50 Return"
        ]
    )

    market_return_shock = (
        market_return
        - SCENARIO_BASE_MARKET_RETURN
    )

    for ticker in (
        scenario_stock_statistics.index
    ):

        expected_return = (
            scenario_stock_statistics.loc[
                ticker,
                "Shrunk Expected Return",
            ]
        )

        stock_beta = (
            scenario_stock_statistics.loc[
                ticker,
                "Beta vs Nifty 50",
            ]
        )

        scenario_return = (
            expected_return
            + stock_beta
            * market_return_shock
        )

        # Prevent extreme model extrapolation.
        scenario_return = float(
            np.clip(
                scenario_return,
                -0.60,
                0.80,
            )
        )

        scenario_asset_return_records.append(
            {
                "Scenario":
                    scenario_name,

                "Ticker":
                    ticker,

                "Nifty 50 Return":
                    market_return,

                "Market Return Shock":
                    market_return_shock,

                "Beta vs Nifty 50":
                    stock_beta,

                "Scenario Asset Return":
                    scenario_return,
            }
        )


scenario_asset_returns_table = (
    pd.DataFrame(
        scenario_asset_return_records
    )
)


# ---------------------------------------------------------
# 5. Calculate portfolio-level scenario outcomes
# ---------------------------------------------------------

scenario_portfolio_records = []

for scenario_name, assumptions in (
    SCENARIO_ASSUMPTIONS.items()
):

    scenario_stock_returns = (
        scenario_asset_returns_table.loc[
            scenario_asset_returns_table[
                "Scenario"
            ]
            == scenario_name
        ]
        .set_index("Ticker")[
            "Scenario Asset Return"
        ]
        .reindex(
            portfolio_returns.columns
        )
    )

    volatility_multiplier = (
        assumptions[
            "Volatility Multiplier"
        ]
    )

    for portfolio_name in (
        latest_portfolio_weights.columns
    ):

        portfolio_weights_for_scenario = (
            latest_portfolio_weights[
                portfolio_name
            ]
        )

        scenario_portfolio_return = float(
            portfolio_weights_for_scenario
            @ scenario_stock_returns
        )

        base_portfolio_volatility = (
            model_portfolio_volatility(
                portfolio_weights_for_scenario,
                simulation_covariance,
            )
        )

        stressed_volatility = (
            base_portfolio_volatility
            * volatility_multiplier
        )

        scenario_ending_value = (
            SCENARIO_INITIAL_VALUE_INR
            * (
                1.0
                + scenario_portfolio_return
            )
        )

        scenario_gain_loss = (
            scenario_ending_value
            - SCENARIO_INITIAL_VALUE_INR
        )

        downside_vs_nifty = (
            scenario_portfolio_return
            - assumptions[
                "Nifty 50 Return"
            ]
        )

        # A simple stressed downside threshold, not formal VaR.
        stressed_downside_threshold = (
            scenario_portfolio_return
            - 1.645
            * stressed_volatility
        )

        stressed_downside_value = (
            SCENARIO_INITIAL_VALUE_INR
            * (
                1.0
                + stressed_downside_threshold
            )
        )

        scenario_portfolio_records.append(
            {
                "Scenario":
                    scenario_name,

                "Portfolio":
                    portfolio_name,

                "Nifty 50 Return":
                    assumptions[
                        "Nifty 50 Return"
                    ],

                "Portfolio Scenario Return":
                    scenario_portfolio_return,

                "Return Difference vs Nifty":
                    downside_vs_nifty,

                "Base Model Volatility":
                    base_portfolio_volatility,

                "Volatility Multiplier":
                    volatility_multiplier,

                "Stressed Volatility":
                    stressed_volatility,

                "Scenario Ending Value (₹)":
                    scenario_ending_value,

                "Scenario Gain/Loss (₹)":
                    scenario_gain_loss,

                "Stressed Downside Threshold":
                    stressed_downside_threshold,

                "Stressed Downside Value (₹)":
                    stressed_downside_value,
            }
        )


portfolio_scenario_results = (
    pd.DataFrame(
        scenario_portfolio_records
    )
)


# ---------------------------------------------------------
# 6. Build comparison tables
# ---------------------------------------------------------

scenario_return_comparison = (
    portfolio_scenario_results
    .pivot(
        index="Portfolio",
        columns="Scenario",
        values="Portfolio Scenario Return",
    )
    .reindex(
        columns=[
            "Bear",
            "Base",
            "Bull",
        ]
    )
)

scenario_ending_value_comparison = (
    portfolio_scenario_results
    .pivot(
        index="Portfolio",
        columns="Scenario",
        values="Scenario Ending Value (₹)",
    )
    .reindex(
        columns=[
            "Bear",
            "Base",
            "Bull",
        ]
    )
)

scenario_stressed_volatility = (
    portfolio_scenario_results
    .pivot(
        index="Portfolio",
        columns="Scenario",
        values="Stressed Volatility",
    )
    .reindex(
        columns=[
            "Bear",
            "Base",
            "Bull",
        ]
    )
)


# ---------------------------------------------------------
# 7. Identify scenario winners
# ---------------------------------------------------------

scenario_winner_records = []

for scenario_name in [
    "Bear",
    "Base",
    "Bull",
]:

    scenario_data = (
        portfolio_scenario_results.loc[
            portfolio_scenario_results[
                "Scenario"
            ]
            == scenario_name
        ]
        .copy()
    )

    best_return_row = (
        scenario_data
        .sort_values(
            "Portfolio Scenario Return",
            ascending=False,
        )
        .iloc[0]
    )

    lowest_risk_row = (
        scenario_data
        .sort_values(
            "Stressed Volatility",
            ascending=True,
        )
        .iloc[0]
    )

    scenario_winner_records.append(
        {
            "Scenario":
                scenario_name,

            "Highest-Return Portfolio":
                best_return_row[
                    "Portfolio"
                ],

            "Highest Portfolio Return":
                best_return_row[
                    "Portfolio Scenario Return"
                ],

            "Lowest-Risk Portfolio":
                lowest_risk_row[
                    "Portfolio"
                ],

            "Lowest Stressed Volatility":
                lowest_risk_row[
                    "Stressed Volatility"
                ],
        }
    )


scenario_winners = (
    pd.DataFrame(
        scenario_winner_records
    )
    .set_index("Scenario")
)


# ---------------------------------------------------------
# 8. Build scenario assumption table
# ---------------------------------------------------------

scenario_assumption_table = (
    pd.DataFrame(
        SCENARIO_ASSUMPTIONS
    )
    .T
)

scenario_assumption_table.index.name = (
    "Scenario"
)


# ---------------------------------------------------------
# 9. Validation
# ---------------------------------------------------------

assert (
    portfolio_scenario_results[
        "Scenario"
    ].nunique()
    == 3
)

assert (
    portfolio_scenario_results[
        "Portfolio"
    ].nunique()
    == 4
)

assert len(
    portfolio_scenario_results
) == 12

assert not (
    portfolio_scenario_results[
        "Portfolio Scenario Return"
    ]
    .isna()
    .any()
)

assert (
    scenario_return_comparison[
        "Bull"
    ]
    >= scenario_return_comparison[
        "Base"
    ]
).all()

assert (
    scenario_return_comparison[
        "Base"
    ]
    >= scenario_return_comparison[
        "Bear"
    ]
).all()


# ---------------------------------------------------------
# 10. Results
# ---------------------------------------------------------

print("Bull, base and bear scenario analysis completed.")
print("-" * 72)
print(
    "Estimation period:",
    scenario_estimation_data.index.min().date(),
    "to",
    scenario_estimation_data.index.max().date(),
)
print(
    "Scenario method:",
    (
        "Shrunk expected returns adjusted by "
        "stock beta and Nifty 50 market shocks"
    ),
)
print(
    "Initial portfolio value:",
    f"₹{SCENARIO_INITIAL_VALUE_INR:,.0f}",
)
print(
    "Stock-return limits:",
    "-60% to +80%",
)
print(
    "Important:",
    (
        "Scenario outputs are assumption-based estimates, "
        "not forecasts"
    ),
)

print("\nSCENARIO ASSUMPTIONS")

display(
    scenario_assumption_table.style.format(
        {
            "Nifty 50 Return":
                "{:.2%}",

            "Volatility Multiplier":
                "{:.2f}x",
        }
    )
)

print("\nPORTFOLIO RETURNS BY SCENARIO")

display(
    scenario_return_comparison.style.format(
        "{:.2%}"
    )
)

print("\nPORTFOLIO ENDING VALUES BY SCENARIO")

display(
    scenario_ending_value_comparison.style.format(
        "₹{:,.0f}"
    )
)

print("\nSTRESSED VOLATILITY BY SCENARIO")

display(
    scenario_stressed_volatility.style.format(
        "{:.2%}"
    )
)

print("\nSCENARIO WINNERS")

display(
    scenario_winners.style.format(
        {
            "Highest Portfolio Return":
                "{:.2%}",

            "Lowest Stressed Volatility":
                "{:.2%}",
        }
    )
)
```

```python
# =========================================================
# USER PORTFOLIO INPUT ENGINE
# Weights | Invested amounts | Quantities
# NSE and BSE listed equities
# =========================================================

def standardise_indian_ticker(
    ticker,
    exchange="NSE",
):
    """
    Convert a user-entered symbol into Yahoo Finance format.

    NSE examples:
        HDFCBANK -> HDFCBANK.NS
        HDFCBANK.NS -> HDFCBANK.NS

    BSE examples:
        500180 -> 500180.BO
        500180.BO -> 500180.BO
    """

    clean_ticker = str(
        ticker
    ).strip().upper()

    clean_exchange = str(
        exchange
    ).strip().upper()

    if not clean_ticker:
        raise ValueError(
            "Ticker cannot be blank."
        )

    if clean_ticker.endswith(
        (
            ".NS",
            ".BO",
        )
    ):
        return clean_ticker

    if clean_exchange == "NSE":
        return f"{clean_ticker}.NS"

    if clean_exchange == "BSE":
        return f"{clean_ticker}.BO"

    raise ValueError(
        f"Unsupported exchange: {exchange}. "
        "Use NSE or BSE."
    )


def download_latest_stock_prices(
    tickers,
):
    """
    Download the latest available adjusted closing price
    for arbitrary NSE/BSE stocks.
    """

    unique_tickers = list(
        dict.fromkeys(
            tickers
        )
    )

    raw_prices = yf.download(
        tickers=unique_tickers,
        period="1mo",
        auto_adjust=True,
        progress=False,
        group_by="column",
        threads=True,
    )

    if raw_prices.empty:
        raise RuntimeError(
            "No recent prices were downloaded."
        )

    if isinstance(
        raw_prices.columns,
        pd.MultiIndex,
    ):

        first_level = (
            raw_prices.columns
            .get_level_values(0)
        )

        second_level = (
            raw_prices.columns
            .get_level_values(1)
        )

        if "Close" in first_level:

            closing_prices = (
                raw_prices["Close"]
                .copy()
            )

        elif "Close" in second_level:

            closing_prices = (
                raw_prices.xs(
                    "Close",
                    axis=1,
                    level=1,
                )
                .copy()
            )

        else:

            raise RuntimeError(
                "Closing-price data were not returned."
            )

    else:

        if "Close" not in raw_prices.columns:

            raise RuntimeError(
                "Closing-price data were not returned."
            )

        closing_prices = pd.DataFrame(
            {
                unique_tickers[0]:
                    raw_prices["Close"]
            }
        )

    latest_price_records = []

    for ticker in unique_tickers:

        if ticker not in closing_prices.columns:

            raise ValueError(
                f"No recent price was found for {ticker}."
            )

        ticker_prices = (
            closing_prices[
                ticker
            ]
            .dropna()
        )

        if ticker_prices.empty:

            raise ValueError(
                f"No valid recent price was found for {ticker}."
            )

        latest_price_records.append(
            {
                "Ticker":
                    ticker,

                "Latest Price (₹)":
                    float(
                        ticker_prices.iloc[-1]
                    ),

                "Price Date":
                    pd.Timestamp(
                        ticker_prices.index[-1]
                    ),
            }
        )

    return (
        pd.DataFrame(
            latest_price_records
        )
        .set_index("Ticker")
    )


def construct_user_portfolio(
    portfolio_input,
    input_mode,
):
    """
    Convert user-entered portfolio data into normalised weights.

    Required columns:
        Ticker
        Value

    Optional column:
        Exchange

    Supported modes:
        Weight
        Invested Amount
        Quantity
    """

    if not isinstance(
        portfolio_input,
        pd.DataFrame,
    ):
        raise TypeError(
            "Portfolio input must be a pandas DataFrame."
        )

    required_columns = {
        "Ticker",
        "Value",
    }

    missing_columns = (
        required_columns
        - set(
            portfolio_input.columns
        )
    )

    if missing_columns:

        raise ValueError(
            "Missing columns: "
            + ", ".join(
                sorted(
                    missing_columns
                )
            )
        )

    portfolio_table = (
        portfolio_input.copy()
    )

    if "Exchange" not in portfolio_table.columns:

        portfolio_table[
            "Exchange"
        ] = "NSE"

    portfolio_table[
        "Standardised Ticker"
    ] = [
        standardise_indian_ticker(
            ticker=ticker,
            exchange=exchange,
        )
        for ticker, exchange in zip(
            portfolio_table[
                "Ticker"
            ],
            portfolio_table[
                "Exchange"
            ],
        )
    ]

    if portfolio_table[
        "Standardised Ticker"
    ].duplicated().any():

        duplicated_tickers = (
            portfolio_table.loc[
                portfolio_table[
                    "Standardised Ticker"
                ].duplicated(
                    keep=False
                ),
                "Standardised Ticker",
            ]
            .unique()
            .tolist()
        )

        raise ValueError(
            "Duplicate holdings found: "
            + ", ".join(
                duplicated_tickers
            )
        )

    portfolio_table[
        "Value"
    ] = pd.to_numeric(
        portfolio_table[
            "Value"
        ],
        errors="coerce",
    )

    if portfolio_table[
        "Value"
    ].isna().any():

        raise ValueError(
            "All portfolio values must be numeric."
        )

    if (
        portfolio_table[
            "Value"
        ]
        <= 0
    ).any():

        raise ValueError(
            "All portfolio values must be greater than zero."
        )

    if len(
        portfolio_table
    ) < 2:

        raise ValueError(
            "Select at least two listed stocks."
        )

    supported_modes = {
        "WEIGHT":
            "Weight",

        "INVESTED AMOUNT":
            "Invested Amount",

        "QUANTITY":
            "Quantity",
    }

    normalised_mode = str(
        input_mode
    ).strip().upper()

    if normalised_mode not in supported_modes:

        raise ValueError(
            "Input mode must be Weight, "
            "Invested Amount or Quantity."
        )

    selected_mode = (
        supported_modes[
            normalised_mode
        ]
    )

    latest_prices = None

    if selected_mode == "Weight":

        portfolio_table[
            "Market Value (₹)"
        ] = np.nan

        portfolio_table[
            "Calculated Weight"
        ] = (
            portfolio_table[
                "Value"
            ]
            / portfolio_table[
                "Value"
            ].sum()
        )

    elif selected_mode == "Invested Amount":

        portfolio_table[
            "Market Value (₹)"
        ] = portfolio_table[
            "Value"
        ]

        portfolio_table[
            "Calculated Weight"
        ] = (
            portfolio_table[
                "Market Value (₹)"
            ]
            / portfolio_table[
                "Market Value (₹)"
            ].sum()
        )

    else:

        latest_prices = (
            download_latest_stock_prices(
                portfolio_table[
                    "Standardised Ticker"
                ].tolist()
            )
        )

        portfolio_table = (
            portfolio_table
            .merge(
                latest_prices.reset_index(),
                left_on="Standardised Ticker",
                right_on="Ticker",
                how="left",
                suffixes=(
                    "",
                    "_Price",
                ),
            )
        )

        portfolio_table.drop(
            columns=[
                "Ticker_Price",
            ],
            errors="ignore",
            inplace=True,
        )

        portfolio_table[
            "Quantity"
        ] = portfolio_table[
            "Value"
        ]

        portfolio_table[
            "Market Value (₹)"
        ] = (
            portfolio_table[
                "Quantity"
            ]
            * portfolio_table[
                "Latest Price (₹)"
            ]
        )

        portfolio_table[
            "Calculated Weight"
        ] = (
            portfolio_table[
                "Market Value (₹)"
            ]
            / portfolio_table[
                "Market Value (₹)"
            ].sum()
        )

    portfolio_weights = (
        portfolio_table
        .set_index(
            "Standardised Ticker"
        )[
            "Calculated Weight"
        ]
        .rename(
            "Portfolio Weight"
        )
    )

    total_market_value = (
        portfolio_table[
            "Market Value (₹)"
        ].sum()
        if portfolio_table[
            "Market Value (₹)"
        ].notna().any()
        else np.nan
    )

    portfolio_summary = pd.Series(
        {
            "Input Mode":
                selected_mode,

            "Number of Holdings":
                len(
                    portfolio_weights
                ),

            "Total Weight":
                portfolio_weights.sum(),

            "Largest Holding":
                portfolio_weights.max(),

            "Top-3 Concentration":
                portfolio_weights.nlargest(
                    3
                ).sum(),

            "Effective Holdings":
                (
                    1.0
                    / np.square(
                        portfolio_weights
                    ).sum()
                ),

            "Total Market Value (₹)":
                total_market_value,
        },
        name="User Portfolio",
    )

    if not np.isclose(
        portfolio_weights.sum(),
        1.0,
        atol=1e-10,
    ):
        raise RuntimeError(
            "Calculated portfolio weights do not sum to 100%."
        )

    return {
        "weights":
            portfolio_weights,

        "portfolio_table":
            portfolio_table,

        "summary":
            portfolio_summary,

        "latest_prices":
            latest_prices,
    }


# ---------------------------------------------------------
# Test 1 — Weight-based portfolio
# ---------------------------------------------------------

weight_input_example = pd.DataFrame(
    {
        "Ticker": [
            "HDFCBANK",
            "TCS",
            "SUNPHARMA",
        ],

        "Exchange": [
            "NSE",
            "NSE",
            "NSE",
        ],

        "Value": [
            40,
            35,
            25,
        ],
    }
)

weight_based_portfolio = (
    construct_user_portfolio(
        portfolio_input=(
            weight_input_example
        ),
        input_mode="Weight",
    )
)


# ---------------------------------------------------------
# Test 2 — Invested-amount portfolio
# ---------------------------------------------------------

amount_input_example = pd.DataFrame(
    {
        "Ticker": [
            "HDFCBANK",
            "TCS",
            "SUNPHARMA",
        ],

        "Exchange": [
            "NSE",
            "NSE",
            "NSE",
        ],

        "Value": [
            400_000,
            350_000,
            250_000,
        ],
    }
)

amount_based_portfolio = (
    construct_user_portfolio(
        portfolio_input=(
            amount_input_example
        ),
        input_mode="Invested Amount",
    )
)


# ---------------------------------------------------------
# Test 3 — Quantity-based portfolio
# ---------------------------------------------------------

quantity_input_example = pd.DataFrame(
    {
        "Ticker": [
            "HDFCBANK",
            "TCS",
            "SUNPHARMA",
        ],

        "Exchange": [
            "NSE",
            "NSE",
            "NSE",
        ],

        "Value": [
            100,
            50,
            75,
        ],
    }
)

quantity_based_portfolio = (
    construct_user_portfolio(
        portfolio_input=(
            quantity_input_example
        ),
        input_mode="Quantity",
    )
)


# ---------------------------------------------------------
# Validation
# ---------------------------------------------------------

for test_portfolio in [
    weight_based_portfolio,
    amount_based_portfolio,
    quantity_based_portfolio,
]:

    assert np.isclose(
        test_portfolio[
            "weights"
        ].sum(),
        1.0,
    )

    assert (
        len(
            test_portfolio[
                "weights"
            ]
        )
        == 3
    )


# ---------------------------------------------------------
# Results
# ---------------------------------------------------------

print("User portfolio input engine completed.")
print("-" * 72)
print(
    "Supported exchanges:",
    "NSE and BSE",
)
print(
    "Supported input modes:",
    "Weight, Invested Amount and Quantity",
)
print(
    "Minimum holdings:",
    2,
)
print(
    "Ticker standardisation:",
    "Automatic .NS and .BO suffixes",
)
print(
    "Quantity valuation:",
    "Latest available adjusted closing price",
)

print("\nWEIGHT-BASED EXAMPLE")

display(
    weight_based_portfolio[
        "portfolio_table"
    ][
        [
            "Standardised Ticker",
            "Value",
            "Calculated Weight",
        ]
    ]
    .style.format(
        {
            "Value": "{:,.2f}",
            "Calculated Weight": "{:.2%}",
        }
    )
)

print("\nINVESTED-AMOUNT EXAMPLE")

display(
    amount_based_portfolio[
        "portfolio_table"
    ][
        [
            "Standardised Ticker",
            "Market Value (₹)",
            "Calculated Weight",
        ]
    ]
    .style.format(
        {
            "Market Value (₹)": "₹{:,.2f}",
            "Calculated Weight": "{:.2%}",
        }
    )
)

print("\nQUANTITY-BASED EXAMPLE")

display(
    quantity_based_portfolio[
        "portfolio_table"
    ][
        [
            "Standardised Ticker",
            "Quantity",
            "Latest Price (₹)",
            "Market Value (₹)",
            "Calculated Weight",
        ]
    ]
    .style.format(
        {
            "Quantity": "{:,.2f}",
            "Latest Price (₹)": "₹{:,.2f}",
            "Market Value (₹)": "₹{:,.2f}",
            "Calculated Weight": "{:.2%}",
        }
    )
)
```

```python
# =========================================================
# GENERIC USER-PORTFOLIO OPTIMISATION ENGINE
# Works with any feasible NSE/BSE portfolio
# =========================================================

def optimise_arbitrary_portfolio(
    asset_returns,
    current_weights,
    benchmark_return_series,
    maximum_stock_weight=0.30,
    minimum_stock_weight=0.00,
    expected_return_shrinkage=0.50,
    risk_free_rate=RISK_FREE_RATE,
):
    """
    Create current, minimum-volatility, maximum-Sharpe and
    risk-parity portfolios for any user-selected stock universe.
    """

    clean_returns = (
        asset_returns
        .replace(
            [np.inf, -np.inf],
            np.nan,
        )
        .dropna()
        .copy()
    )

    asset_names_local = (
        clean_returns.columns.tolist()
    )

    number_of_assets = len(
        asset_names_local
    )

    if number_of_assets < 2:
        raise ValueError(
            "At least two stocks are required."
        )

    current_weight_series = (
        pd.Series(
            current_weights,
            dtype=float,
        )
        .reindex(
            asset_names_local
        )
    )

    if current_weight_series.isna().any():
        raise ValueError(
            "Current weights do not match the return columns."
        )

    current_weight_series = (
        current_weight_series
        / current_weight_series.sum()
    )

    if (
        number_of_assets
        * maximum_stock_weight
        < 1.0 - 1e-10
    ):
        minimum_feasible_cap = (
            1.0
            / number_of_assets
        )

        raise ValueError(
            f"A {maximum_stock_weight:.0%} cap is infeasible for "
            f"{number_of_assets} stocks. The cap must be at least "
            f"{minimum_feasible_cap:.2%}."
        )

    if (
        number_of_assets
        * minimum_stock_weight
        > 1.0 + 1e-10
    ):
        raise ValueError(
            "The minimum-weight constraint is infeasible."
        )

    # -----------------------------------------------------
    # Estimate expected returns
    # -----------------------------------------------------

    historical_expected_returns_local = (
        clean_returns.mean()
        * TRADING_DAYS_PER_YEAR
    )

    average_expected_return = (
        historical_expected_returns_local.mean()
    )

    shrunk_expected_returns_local = (
        (
            1.0
            - expected_return_shrinkage
        )
        * historical_expected_returns_local
        + expected_return_shrinkage
        * average_expected_return
    )

    # -----------------------------------------------------
    # Estimate robust covariance
    # -----------------------------------------------------

    covariance_estimator = (
        LedoitWolf()
        .fit(
            clean_returns.to_numpy()
        )
    )

    covariance_local = (
        pd.DataFrame(
            covariance_estimator.covariance_,
            index=asset_names_local,
            columns=asset_names_local,
        )
        * TRADING_DAYS_PER_YEAR
    )

    covariance_array = (
        covariance_local.to_numpy()
    )

    expected_return_array = (
        shrunk_expected_returns_local.to_numpy()
    )

    # Equal weight is always a feasible starting point
    # after the constraint checks above.
    starting_weights = np.repeat(
        1.0 / number_of_assets,
        number_of_assets,
    )

    bounds = [
        (
            minimum_stock_weight,
            maximum_stock_weight,
        )
        for _ in range(
            number_of_assets
        )
    ]

    full_investment_constraint_local = {
        "type": "eq",
        "fun": lambda weights: (
            np.sum(weights) - 1.0
        ),
    }

    # -----------------------------------------------------
    # Local objective functions
    # -----------------------------------------------------

    def portfolio_return_objective(
        weights,
    ):
        return float(
            weights
            @ expected_return_array
        )

    def portfolio_volatility_objective(
        weights,
    ):
        variance = float(
            weights.T
            @ covariance_array
            @ weights
        )

        return float(
            np.sqrt(
                max(
                    variance,
                    0.0,
                )
            )
        )

    def negative_sharpe_local(
        weights,
    ):
        volatility = (
            portfolio_volatility_objective(
                weights
            )
        )

        if volatility <= 0:
            return 1e10

        expected_return = (
            portfolio_return_objective(
                weights
            )
        )

        return -(
            (
                expected_return
                - risk_free_rate
            )
            / volatility
        )

    def risk_parity_local(
        weights,
    ):
        marginal_variance = (
            covariance_array
            @ weights
        )

        total_variance = float(
            weights.T
            @ covariance_array
            @ weights
        )

        if total_variance <= 0:
            return 1e10

        percentage_risk_contribution = (
            weights
            * marginal_variance
            / total_variance
        )

        target_contribution = (
            1.0
            / number_of_assets
        )

        return float(
            np.sum(
                (
                    percentage_risk_contribution
                    - target_contribution
                )
                ** 2
            )
        )

    # -----------------------------------------------------
    # Generic solver
    # -----------------------------------------------------

    def solve_portfolio(
        portfolio_name,
        objective_function,
    ):
        result = minimize(
            fun=objective_function,
            x0=starting_weights,
            method="SLSQP",
            bounds=bounds,
            constraints=[
                full_investment_constraint_local
            ],
            options={
                "maxiter": 3000,
                "ftol": 1e-12,
                "disp": False,
            },
        )

        if not result.success:
            raise RuntimeError(
                f"{portfolio_name} optimisation failed: "
                f"{result.message}"
            )

        solved_weights = pd.Series(
            result.x,
            index=asset_names_local,
            name=portfolio_name,
        )

        solved_weights[
            solved_weights.abs() < 1e-8
        ] = 0.0

        solved_weights = (
            solved_weights
            / solved_weights.sum()
        )

        return solved_weights

    minimum_volatility_weights_local = (
        solve_portfolio(
            "Minimum Volatility",
            portfolio_volatility_objective,
        )
    )

    maximum_sharpe_weights_local = (
        solve_portfolio(
            "Maximum Sharpe",
            negative_sharpe_local,
        )
    )

    risk_parity_weights_local = (
        solve_portfolio(
            "Risk Parity",
            risk_parity_local,
        )
    )

    portfolio_weight_sets_local = {
        "Current Portfolio":
            current_weight_series,

        "Minimum Volatility":
            minimum_volatility_weights_local,

        "Maximum Sharpe":
            maximum_sharpe_weights_local,

        "Risk Parity":
            risk_parity_weights_local,
    }

    weight_comparison = pd.DataFrame(
        portfolio_weight_sets_local
    )

    # -----------------------------------------------------
    # Model-implied estimates
    # -----------------------------------------------------

    model_records = []

    for portfolio_name, weights in (
        portfolio_weight_sets_local.items()
    ):

        weight_array = (
            weights.to_numpy()
        )

        expected_return = float(
            weight_array
            @ expected_return_array
        )

        volatility = float(
            np.sqrt(
                weight_array.T
                @ covariance_array
                @ weight_array
            )
        )

        sharpe_ratio = (
            (
                expected_return
                - risk_free_rate
            )
            / volatility
            if volatility > 0
            else np.nan
        )

        risk_contributions = (
            calculate_percentage_risk_contributions(
                weights,
                covariance_local,
            )
        )

        model_records.append(
            {
                "Portfolio":
                    portfolio_name,

                "Expected Return":
                    expected_return,

                "Volatility":
                    volatility,

                "Sharpe Ratio":
                    sharpe_ratio,

                "Largest Weight":
                    weights.max(),

                "Active Holdings":
                    int(
                        (
                            weights > 0.001
                        ).sum()
                    ),

                "Largest Risk Contribution":
                    np.nanmax(
                        risk_contributions
                    ),
            }
        )

    model_comparison = (
        pd.DataFrame(
            model_records
        )
        .set_index("Portfolio")
    )

    # -----------------------------------------------------
    # Realised historical comparison
    # -----------------------------------------------------

    realised_records = {}

    for portfolio_name, weights in (
        portfolio_weight_sets_local.items()
    ):

        _, statistics = (
            calculate_portfolio_statistics(
                asset_returns=clean_returns,
                weights=weights,
                benchmark_return_series=(
                    benchmark_return_series
                ),
                risk_free_rate=(
                    risk_free_rate
                ),
            )
        )

        realised_records[
            portfolio_name
        ] = statistics

    realised_comparison = (
        pd.DataFrame(
            realised_records
        )
        .T
    )

    # -----------------------------------------------------
    # Validation
    # -----------------------------------------------------

    assert np.allclose(
        weight_comparison.sum(axis=0),
        1.0,
        atol=1e-6,
    )

    assert (
        weight_comparison[
            [
                "Minimum Volatility",
                "Maximum Sharpe",
                "Risk Parity",
            ]
        ]
        .max()
        .max()
        <= maximum_stock_weight + 1e-6
    )

    return {
        "weights":
            weight_comparison,

        "model_comparison":
            model_comparison,

        "realised_comparison":
            realised_comparison,

        "expected_returns":
            shrunk_expected_returns_local,

        "covariance":
            covariance_local,

        "constraints":
            {
                "Minimum Weight":
                    minimum_stock_weight,

                "Maximum Weight":
                    maximum_stock_weight,

                "Holdings":
                    number_of_assets,
            },
    }


# =========================================================
# TEST THE COMPLETE USER-PORTFOLIO PIPELINE
# =========================================================

custom_portfolio_input = pd.DataFrame(
    {
        "Ticker": [
            "HDFCBANK",
            "TCS",
            "SUNPHARMA",
            "BHARTIARTL",
            "LT",
            "M&M",
        ],

        "Exchange": [
            "NSE",
            "NSE",
            "NSE",
            "NSE",
            "NSE",
            "NSE",
        ],

        "Value": [
            250_000,
            150_000,
            200_000,
            150_000,
            150_000,
            100_000,
        ],
    }
)

custom_portfolio = (
    construct_user_portfolio(
        portfolio_input=(
            custom_portfolio_input
        ),
        input_mode="Invested Amount",
    )
)

custom_portfolio_weights = (
    custom_portfolio[
        "weights"
    ]
)

custom_market_data = (
    download_portfolio_market_data(
        portfolio_weights=(
            custom_portfolio_weights.to_dict()
        ),
        start_date=DEFAULT_START_DATE,
        end_date=DEFAULT_END_DATE,
        benchmark=DEFAULT_BENCHMARK,
    )
)

custom_optimisation_results = (
    optimise_arbitrary_portfolio(
        asset_returns=(
            custom_market_data[
                "portfolio_returns"
            ]
        ),
        current_weights=(
            custom_market_data[
                "weights"
            ]
        ),
        benchmark_return_series=(
            custom_market_data[
                "benchmark_returns"
            ]
        ),
        maximum_stock_weight=0.30,
        minimum_stock_weight=0.00,
    )
)


print("Generic user-portfolio optimisation completed.")
print("-" * 72)
print(
    "Custom holdings:",
    len(
        custom_portfolio_weights
    ),
)
print(
    "Input mode:",
    custom_portfolio[
        "summary"
    ][
        "Input Mode"
    ],
)
print(
    "Analysis period:",
    custom_market_data[
        "portfolio_returns"
    ].index.min().date(),
    "to",
    custom_market_data[
        "portfolio_returns"
    ].index.max().date(),
)
print(
    "Maximum stock weight:",
    f"{custom_optimisation_results['constraints']['Maximum Weight']:.0%}",
)
print(
    "Optimised portfolios:",
    "Minimum Volatility, Maximum Sharpe and Risk Parity",
)

print("\nCUSTOM PORTFOLIO WEIGHTS")

display(
    custom_optimisation_results[
        "weights"
    ].style.format(
        "{:.2%}"
    )
)

print("\nMODEL-IMPLIED COMPARISON")

display(
    custom_optimisation_results[
        "model_comparison"
    ].style.format(
        {
            "Expected Return": "{:.2%}",
            "Volatility": "{:.2%}",
            "Sharpe Ratio": "{:.2f}",
            "Largest Weight": "{:.2%}",
            "Active Holdings": "{:.0f}",
            "Largest Risk Contribution": "{:.2%}",
        }
    )
)

print("\nREALISED HISTORICAL COMPARISON")

display(
    custom_optimisation_results[
        "realised_comparison"
    ][
        [
            "Ending Value (₹)",
            "CAGR",
            "Annualised Volatility",
            "Sharpe Ratio",
            "Maximum Drawdown",
        ]
    ].style.format(
        {
            "Ending Value (₹)": "₹{:,.0f}",
            "CAGR": "{:.2%}",
            "Annualised Volatility": "{:.2%}",
            "Sharpe Ratio": "{:.2f}",
            "Maximum Drawdown": "{:.2%}",
        }
    )
)
```

```python
# =========================================================
# GENERIC USER-PORTFOLIO RISK ENGINE
# Concentration | VaR | Expected Shortfall | Monte Carlo
# Bull, Base and Bear Scenarios
# =========================================================

def analyse_arbitrary_portfolio_risk(
    asset_returns,
    benchmark_returns,
    portfolio_weights,
    expected_returns,
    covariance_matrix,
    initial_value_inr=INITIAL_INVESTMENT_INR,
    simulations=10_000,
    horizon_days=252,
    block_length=21,
    estimation_window=756,
    random_seed=42,
):
    """
    Run a complete risk analysis for arbitrary user-selected portfolios.

    portfolio_weights:
        DataFrame with tickers as rows and portfolio alternatives
        as columns.
    """

    # -----------------------------------------------------
    # 1. Align and validate inputs
    # -----------------------------------------------------

    clean_asset_returns = (
        asset_returns
        .replace(
            [np.inf, -np.inf],
            np.nan,
        )
        .dropna()
        .copy()
    )

    clean_weights = (
        portfolio_weights
        .reindex(
            index=clean_asset_returns.columns
        )
        .astype(float)
    )

    if clean_weights.isna().any().any():
        raise ValueError(
            "Portfolio weights do not match the return columns."
        )

    if not np.allclose(
        clean_weights.sum(axis=0),
        1.0,
        atol=1e-6,
    ):
        raise ValueError(
            "Every portfolio must sum to 100%."
        )

    aligned_expected_returns = (
        pd.Series(
            expected_returns,
            dtype=float,
        )
        .reindex(
            clean_asset_returns.columns
        )
    )

    aligned_covariance = (
        covariance_matrix
        .reindex(
            index=clean_asset_returns.columns,
            columns=clean_asset_returns.columns,
        )
    )

    portfolio_names = (
        clean_weights.columns.tolist()
    )

    weight_matrix = (
        clean_weights.to_numpy()
    )

    # -----------------------------------------------------
    # 2. Concentration and diversification analysis
    # -----------------------------------------------------

    diversification_records = {}

    for portfolio_name in portfolio_names:

        diversification_records[
            portfolio_name
        ] = calculate_diversification_statistics(
            weights=clean_weights[
                portfolio_name
            ],
            annualised_covariance=(
                aligned_covariance
            ),
        )

    user_diversification_summary = (
        pd.DataFrame(
            diversification_records
        )
        .T
    )

    # -----------------------------------------------------
    # 3. Historical daily portfolio returns
    # -----------------------------------------------------

    daily_portfolio_returns = (
        clean_asset_returns
        @ clean_weights
    )

    daily_portfolio_returns.columns = (
        portfolio_names
    )

    # -----------------------------------------------------
    # 4. One-day historical VaR and Expected Shortfall
    # -----------------------------------------------------

    one_day_risk_records = []

    for portfolio_name in portfolio_names:

        return_array = (
            daily_portfolio_returns[
                portfolio_name
            ]
            .to_numpy()
        )

        (
            var_95,
            es_95,
        ) = calculate_var_and_expected_shortfall(
            return_array,
            confidence_level=0.95,
        )

        (
            var_99,
            es_99,
        ) = calculate_var_and_expected_shortfall(
            return_array,
            confidence_level=0.99,
        )

        one_day_risk_records.append(
            {
                "Portfolio":
                    portfolio_name,

                "1-Day VaR 95%":
                    var_95,

                "1-Day ES 95%":
                    es_95,

                "1-Day VaR 99%":
                    var_99,

                "1-Day ES 99%":
                    es_99,

                "Worst Historical Day":
                    return_array.min(),

                "95% VaR Amount (₹)":
                    initial_value_inr
                    * var_95,

                "95% ES Amount (₹)":
                    initial_value_inr
                    * es_95,
            }
        )

    user_one_day_risk = (
        pd.DataFrame(
            one_day_risk_records
        )
        .set_index("Portfolio")
    )

    # -----------------------------------------------------
    # 5. Prepare block-bootstrap estimation history
    # -----------------------------------------------------

    estimation_returns = (
        clean_asset_returns
        .tail(
            estimation_window
        )
        .copy()
    )

    if len(estimation_returns) < 252:
        raise ValueError(
            "At least 252 return observations are required."
        )

    target_daily_returns = (
        aligned_expected_returns
        / TRADING_DAYS_PER_YEAR
    )

    adjusted_bootstrap_returns = (
        estimation_returns
        - estimation_returns.mean()
        + target_daily_returns
    )

    adjusted_return_array = (
        adjusted_bootstrap_returns
        .to_numpy(
            dtype=float
        )
    )

    number_of_blocks = int(
        np.ceil(
            horizon_days
            / block_length
        )
    )

    maximum_block_start = (
        len(adjusted_return_array)
        - block_length
    )

    if maximum_block_start < 1:
        raise ValueError(
            "The estimation period is too short for block bootstrap."
        )

    # -----------------------------------------------------
    # 6. Monte Carlo block-bootstrap simulation
    # -----------------------------------------------------

    random_generator = np.random.default_rng(
        random_seed
    )

    simulation_chunk_size = 500
    completed_simulations = 0

    terminal_return_chunks = []
    maximum_drawdown_chunks = []

    while completed_simulations < simulations:

        current_chunk_size = min(
            simulation_chunk_size,
            simulations
            - completed_simulations,
        )

        block_start_indices = (
            random_generator.integers(
                low=0,
                high=maximum_block_start + 1,
                size=(
                    current_chunk_size,
                    number_of_blocks,
                ),
            )
        )

        block_offsets = np.arange(
            block_length
        )

        sampled_indices = (
            block_start_indices[
                :,
                :,
                None,
            ]
            + block_offsets[
                None,
                None,
                :,
            ]
        )

        sampled_indices = (
            sampled_indices
            .reshape(
                current_chunk_size,
                -1,
            )
            [
                :,
                :horizon_days,
            ]
        )

        simulated_asset_returns = (
            adjusted_return_array[
                sampled_indices
            ]
        )

        simulated_asset_wealth = np.cumprod(
            1.0
            + simulated_asset_returns,
            axis=1,
        )

        simulated_portfolio_wealth = np.einsum(
            "sda,ap->sdp",
            simulated_asset_wealth,
            weight_matrix,
        )

        terminal_returns = (
            simulated_portfolio_wealth[
                :,
                -1,
                :,
            ]
            - 1.0
        )

        running_peaks = np.maximum.accumulate(
            simulated_portfolio_wealth,
            axis=1,
        )

        simulated_drawdowns = (
            simulated_portfolio_wealth
            / running_peaks
            - 1.0
        )

        maximum_drawdowns = (
            simulated_drawdowns.min(
                axis=1
            )
        )

        terminal_return_chunks.append(
            terminal_returns
        )

        maximum_drawdown_chunks.append(
            maximum_drawdowns
        )

        completed_simulations += (
            current_chunk_size
        )

    simulated_terminal_returns = np.vstack(
        terminal_return_chunks
    )

    simulated_maximum_drawdowns = np.vstack(
        maximum_drawdown_chunks
    )

    # -----------------------------------------------------
    # 7. Monte Carlo risk summary
    # -----------------------------------------------------

    monte_carlo_records = []

    for portfolio_number, portfolio_name in enumerate(
        portfolio_names
    ):

        terminal_returns = (
            simulated_terminal_returns[
                :,
                portfolio_number,
            ]
        )

        drawdowns = (
            simulated_maximum_drawdowns[
                :,
                portfolio_number,
            ]
        )

        (
            one_year_var_95,
            one_year_es_95,
        ) = calculate_var_and_expected_shortfall(
            terminal_returns,
            confidence_level=0.95,
        )

        (
            one_year_var_99,
            one_year_es_99,
        ) = calculate_var_and_expected_shortfall(
            terminal_returns,
            confidence_level=0.99,
        )

        monte_carlo_records.append(
            {
                "Portfolio":
                    portfolio_name,

                "Median Terminal Return":
                    np.median(
                        terminal_returns
                    ),

                "5th Percentile Return":
                    np.quantile(
                        terminal_returns,
                        0.05,
                    ),

                "95th Percentile Return":
                    np.quantile(
                        terminal_returns,
                        0.95,
                    ),

                "Probability of Loss":
                    np.mean(
                        terminal_returns < 0
                    ),

                "Probability of Loss > 10%":
                    np.mean(
                        terminal_returns < -0.10
                    ),

                "Probability of Loss > 20%":
                    np.mean(
                        terminal_returns < -0.20
                    ),

                "One-Year VaR 95%":
                    one_year_var_95,

                "One-Year ES 95%":
                    one_year_es_95,

                "One-Year VaR 99%":
                    one_year_var_99,

                "One-Year ES 99%":
                    one_year_es_99,

                "Median Maximum Drawdown":
                    np.median(
                        drawdowns
                    ),

                "5th Percentile Maximum Drawdown":
                    np.quantile(
                        drawdowns,
                        0.05,
                    ),

                "Median Terminal Value (₹)":
                    initial_value_inr
                    * (
                        1.0
                        + np.median(
                            terminal_returns
                        )
                    ),

                "5th Percentile Terminal Value (₹)":
                    initial_value_inr
                    * (
                        1.0
                        + np.quantile(
                            terminal_returns,
                            0.05,
                        )
                    ),
            }
        )

    user_monte_carlo_summary = (
        pd.DataFrame(
            monte_carlo_records
        )
        .set_index("Portfolio")
    )

    # -----------------------------------------------------
    # 8. Bull, base and bear scenarios
    # -----------------------------------------------------

    benchmark_aligned_data = pd.concat(
        [
            estimation_returns,
            benchmark_returns.rename(
                "Nifty 50"
            ),
        ],
        axis=1,
    ).dropna()

    benchmark_series = (
        benchmark_aligned_data[
            "Nifty 50"
        ]
    )

    benchmark_variance = (
        benchmark_series.var(
            ddof=1
        )
    )

    stock_betas = {}

    for ticker in (
        estimation_returns.columns
    ):

        stock_betas[
            ticker
        ] = (
            benchmark_aligned_data[
                ticker
            ]
            .cov(
                benchmark_series
            )
            / benchmark_variance
        )

    stock_beta_series = pd.Series(
        stock_betas
    )

    base_market_return = (
        SCENARIO_ASSUMPTIONS[
            "Base"
        ][
            "Nifty 50 Return"
        ]
    )

    scenario_records = []

    for scenario_name, assumptions in (
        SCENARIO_ASSUMPTIONS.items()
    ):

        market_return_shock = (
            assumptions[
                "Nifty 50 Return"
            ]
            - base_market_return
        )

        scenario_asset_returns = (
            aligned_expected_returns
            + stock_beta_series
            * market_return_shock
        ).clip(
            lower=-0.60,
            upper=0.80,
        )

        for portfolio_name in (
            portfolio_names
        ):

            weights = clean_weights[
                portfolio_name
            ]

            scenario_return = float(
                weights
                @ scenario_asset_returns
            )

            base_volatility = (
                model_portfolio_volatility(
                    weights,
                    aligned_covariance,
                )
            )

            stressed_volatility = (
                base_volatility
                * assumptions[
                    "Volatility Multiplier"
                ]
            )

            scenario_records.append(
                {
                    "Scenario":
                        scenario_name,

                    "Portfolio":
                        portfolio_name,

                    "Nifty 50 Return":
                        assumptions[
                            "Nifty 50 Return"
                        ],

                    "Portfolio Return":
                        scenario_return,

                    "Stressed Volatility":
                        stressed_volatility,

                    "Ending Value (₹)":
                        initial_value_inr
                        * (
                            1.0
                            + scenario_return
                        ),

                    "Gain/Loss (₹)":
                        initial_value_inr
                        * scenario_return,
                }
            )

    user_scenario_results = pd.DataFrame(
        scenario_records
    )

    user_scenario_return_table = (
        user_scenario_results
        .pivot(
            index="Portfolio",
            columns="Scenario",
            values="Portfolio Return",
        )
        .reindex(
            columns=[
                "Bear",
                "Base",
                "Bull",
            ]
        )
    )

    # -----------------------------------------------------
    # 9. Validation
    # -----------------------------------------------------

    assert simulated_terminal_returns.shape == (
        simulations,
        len(portfolio_names),
    )

    assert not np.isnan(
        simulated_terminal_returns
    ).any()

    assert (
        user_scenario_results[
            "Scenario"
        ].nunique()
        == 3
    )

    assert (
        user_scenario_results[
            "Portfolio"
        ].nunique()
        == len(
            portfolio_names
        )
    )

    return {
        "daily_returns":
            daily_portfolio_returns,

        "diversification":
            user_diversification_summary,

        "one_day_risk":
            user_one_day_risk,

        "monte_carlo":
            user_monte_carlo_summary,

        "scenario_results":
            user_scenario_results,

        "scenario_return_table":
            user_scenario_return_table,

        "simulation_terminal_returns":
            simulated_terminal_returns,

        "simulation_drawdowns":
            simulated_maximum_drawdowns,
    }


# =========================================================
# TEST ON THE CUSTOM SIX-STOCK USER PORTFOLIO
# =========================================================

custom_user_risk_results = (
    analyse_arbitrary_portfolio_risk(
        asset_returns=(
            custom_market_data[
                "portfolio_returns"
            ]
        ),
        benchmark_returns=(
            custom_market_data[
                "benchmark_returns"
            ]
        ),
        portfolio_weights=(
            custom_optimisation_results[
                "weights"
            ]
        ),
        expected_returns=(
            custom_optimisation_results[
                "expected_returns"
            ]
        ),
        covariance_matrix=(
            custom_optimisation_results[
                "covariance"
            ]
        ),
        initial_value_inr=1_000_000,
        simulations=10_000,
        horizon_days=252,
        block_length=21,
    )
)


print("Generic user-portfolio risk analysis completed.")
print("-" * 72)
print(
    "Portfolios analysed:",
    len(
        custom_optimisation_results[
            "weights"
        ].columns
    ),
)
print(
    "Stocks in user universe:",
    len(
        custom_optimisation_results[
            "weights"
        ]
    ),
)
print(
    "Monte Carlo simulations:",
    f"{10_000:,}",
)
print(
    "Simulation horizon:",
    "252 trading days",
)
print(
    "Risk modules:",
    (
        "Diversification, VaR, Expected Shortfall, "
        "Monte Carlo and scenarios"
    ),
)

print("\nDIVERSIFICATION SUMMARY")

display(
    custom_user_risk_results[
        "diversification"
    ].style.format(
        {
            "Largest Holding": "{:.2%}",
            "Top-3 Concentration": "{:.2%}",
            "Herfindahl Index": "{:.4f}",
            "Effective Holdings": "{:.2f}",
            "Average Pairwise Correlation": "{:.2f}",
            "Diversification Ratio": "{:.2f}",
        }
    )
)

print("\nONE-DAY VaR AND EXPECTED SHORTFALL")

display(
    custom_user_risk_results[
        "one_day_risk"
    ].style.format(
        {
            "1-Day VaR 95%": "{:.2%}",
            "1-Day ES 95%": "{:.2%}",
            "1-Day VaR 99%": "{:.2%}",
            "1-Day ES 99%": "{:.2%}",
            "Worst Historical Day": "{:.2%}",
            "95% VaR Amount (₹)": "₹{:,.0f}",
            "95% ES Amount (₹)": "₹{:,.0f}",
        }
    )
)

print("\nONE-YEAR MONTE CARLO RISK")

display(
    custom_user_risk_results[
        "monte_carlo"
    ][
        [
            "Median Terminal Return",
            "5th Percentile Return",
            "95th Percentile Return",
            "Probability of Loss",
            "Probability of Loss > 10%",
            "One-Year VaR 95%",
            "One-Year ES 95%",
            "Median Maximum Drawdown",
            "5th Percentile Maximum Drawdown",
            "Median Terminal Value (₹)",
        ]
    ].style.format(
        {
            "Median Terminal Return": "{:.2%}",
            "5th Percentile Return": "{:.2%}",
            "95th Percentile Return": "{:.2%}",
            "Probability of Loss": "{:.2%}",
            "Probability of Loss > 10%": "{:.2%}",
            "One-Year VaR 95%": "{:.2%}",
            "One-Year ES 95%": "{:.2%}",
            "Median Maximum Drawdown": "{:.2%}",
            "5th Percentile Maximum Drawdown": "{:.2%}",
            "Median Terminal Value (₹)": "₹{:,.0f}",
        }
    )
)

print("\nBULL, BASE AND BEAR RETURNS")

display(
    custom_user_risk_results[
        "scenario_return_table"
    ].style.format(
        "{:.2%}"
    )
)
```

```python
# =========================================================
# GENERIC EFFICIENT FRONTIER
# Works for any user-selected NSE/BSE portfolio
# =========================================================

def build_arbitrary_efficient_frontier(
    expected_returns,
    covariance_matrix,
    comparison_weights=None,
    minimum_stock_weight=0.00,
    maximum_stock_weight=0.30,
    risk_free_rate=RISK_FREE_RATE,
    frontier_points=60,
):
    """
    Build a constrained efficient frontier for an arbitrary
    user-selected stock universe.
    """

    expected_return_series = pd.Series(
        expected_returns,
        dtype=float,
    )

    covariance = (
        covariance_matrix
        .reindex(
            index=expected_return_series.index,
            columns=expected_return_series.index,
        )
        .astype(float)
    )

    asset_names_local = (
        expected_return_series.index.tolist()
    )

    number_of_assets = len(
        asset_names_local
    )

    if number_of_assets < 2:
        raise ValueError(
            "At least two stocks are required."
        )

    if (
        number_of_assets
        * maximum_stock_weight
        < 1.0 - 1e-10
    ):
        raise ValueError(
            f"The {maximum_stock_weight:.0%} maximum-weight cap "
            f"is infeasible for {number_of_assets} stocks."
        )

    expected_return_array = (
        expected_return_series.to_numpy()
    )

    covariance_array = (
        covariance.to_numpy()
    )

    starting_weights = np.repeat(
        1.0 / number_of_assets,
        number_of_assets,
    )

    bounds = [
        (
            minimum_stock_weight,
            maximum_stock_weight,
        )
        for _ in range(
            number_of_assets
        )
    ]

    full_investment_constraint_local = {
        "type": "eq",
        "fun": lambda weights: (
            np.sum(weights) - 1.0
        ),
    }

    def portfolio_return_local(
        weights,
    ):
        return float(
            np.asarray(weights)
            @ expected_return_array
        )

    def portfolio_volatility_local(
        weights,
    ):
        weight_array = np.asarray(
            weights,
            dtype=float,
        )

        variance = float(
            weight_array.T
            @ covariance_array
            @ weight_array
        )

        return float(
            np.sqrt(
                max(
                    variance,
                    0.0,
                )
            )
        )

    def solve_weights(
        objective_function,
        constraints,
        initial_guess,
        portfolio_name,
    ):
        result = minimize(
            fun=objective_function,
            x0=np.asarray(
                initial_guess,
                dtype=float,
            ),
            method="SLSQP",
            bounds=bounds,
            constraints=constraints,
            options={
                "maxiter": 4000,
                "ftol": 1e-12,
                "disp": False,
            },
        )

        if not result.success:
            raise RuntimeError(
                f"{portfolio_name} failed: "
                f"{result.message}"
            )

        solved_weights = pd.Series(
            result.x,
            index=asset_names_local,
            name=portfolio_name,
        )

        solved_weights[
            solved_weights.abs() < 1e-8
        ] = 0.0

        solved_weights = (
            solved_weights
            / solved_weights.sum()
        )

        return solved_weights

    # -----------------------------------------------------
    # Minimum-volatility endpoint
    # -----------------------------------------------------

    minimum_volatility_weights_local = (
        solve_weights(
            objective_function=(
                portfolio_volatility_local
            ),
            constraints=[
                full_investment_constraint_local
            ],
            initial_guess=starting_weights,
            portfolio_name="Minimum Volatility",
        )
    )

    minimum_feasible_return = (
        portfolio_return_local(
            minimum_volatility_weights_local
        )
    )

    # -----------------------------------------------------
    # Maximum-return endpoint
    # -----------------------------------------------------

    maximum_return_weights_local = (
        solve_weights(
            objective_function=lambda weights: (
                -portfolio_return_local(
                    weights
                )
            ),
            constraints=[
                full_investment_constraint_local
            ],
            initial_guess=starting_weights,
            portfolio_name="Maximum Feasible Return",
        )
    )

    maximum_feasible_return = (
        portfolio_return_local(
            maximum_return_weights_local
        )
    )

    target_returns = np.linspace(
        minimum_feasible_return,
        maximum_feasible_return - 1e-7,
        frontier_points,
    )

    # -----------------------------------------------------
    # Solve each efficient-frontier point
    # -----------------------------------------------------

    frontier_records = []
    frontier_weight_records = []

    rolling_starting_weights = (
        minimum_volatility_weights_local
        .to_numpy()
    )

    for frontier_number, target_return in enumerate(
        target_returns,
        start=1,
    ):

        target_return_constraint = {
            "type": "eq",
            "fun": (
                lambda weights, target=target_return: (
                    portfolio_return_local(
                        weights
                    )
                    - target
                )
            ),
        }

        result = minimize(
            fun=portfolio_volatility_local,
            x0=rolling_starting_weights,
            method="SLSQP",
            bounds=bounds,
            constraints=[
                full_investment_constraint_local,
                target_return_constraint,
            ],
            options={
                "maxiter": 4000,
                "ftol": 1e-12,
                "disp": False,
            },
        )

        if not result.success:
            continue

        frontier_weights = pd.Series(
            result.x,
            index=asset_names_local,
        )

        frontier_weights[
            frontier_weights.abs() < 1e-8
        ] = 0.0

        frontier_weights = (
            frontier_weights
            / frontier_weights.sum()
        )

        expected_return = (
            portfolio_return_local(
                frontier_weights
            )
        )

        volatility = (
            portfolio_volatility_local(
                frontier_weights
            )
        )

        sharpe_ratio = (
            (
                expected_return
                - risk_free_rate
            )
            / volatility
            if volatility > 0
            else np.nan
        )

        frontier_records.append(
            {
                "Frontier Portfolio":
                    frontier_number,

                "Expected Return":
                    expected_return,

                "Volatility":
                    volatility,

                "Sharpe Ratio":
                    sharpe_ratio,

                "Largest Weight":
                    frontier_weights.max(),

                "Active Holdings":
                    int(
                        (
                            frontier_weights
                            > 0.001
                        ).sum()
                    ),
            }
        )

        weight_record = {
            "Frontier Portfolio":
                frontier_number,
        }

        weight_record.update(
            frontier_weights.to_dict()
        )

        frontier_weight_records.append(
            weight_record
        )

        rolling_starting_weights = (
            frontier_weights.to_numpy()
        )

    efficient_frontier_local = (
        pd.DataFrame(
            frontier_records
        )
        .set_index(
            "Frontier Portfolio"
        )
    )

    efficient_frontier_weights_local = (
        pd.DataFrame(
            frontier_weight_records
        )
        .set_index(
            "Frontier Portfolio"
        )
    )

    if efficient_frontier_local.empty:
        raise RuntimeError(
            "No efficient-frontier portfolios were generated."
        )

    best_frontier_index = (
        efficient_frontier_local[
            "Sharpe Ratio"
        ]
        .idxmax()
    )

    best_frontier_result = (
        efficient_frontier_local.loc[
            best_frontier_index
        ]
    )

    best_frontier_weights = (
        efficient_frontier_weights_local.loc[
            best_frontier_index
        ]
    )

    # -----------------------------------------------------
    # Build comparison points
    # -----------------------------------------------------

    comparison_records = []

    if comparison_weights is not None:

        aligned_comparison_weights = (
            comparison_weights
            .reindex(
                index=asset_names_local
            )
        )

        for portfolio_name in (
            aligned_comparison_weights.columns
        ):

            weights = (
                aligned_comparison_weights[
                    portfolio_name
                ]
            )

            expected_return = (
                portfolio_return_local(
                    weights
                )
            )

            volatility = (
                portfolio_volatility_local(
                    weights
                )
            )

            comparison_records.append(
                {
                    "Portfolio":
                        portfolio_name,

                    "Expected Return":
                        expected_return,

                    "Volatility":
                        volatility,

                    "Sharpe Ratio":
                        (
                            (
                                expected_return
                                - risk_free_rate
                            )
                            / volatility
                        ),
                }
            )

    comparison_points_local = (
        pd.DataFrame(
            comparison_records
        )
        .set_index("Portfolio")
        if comparison_records
        else pd.DataFrame()
    )

    # -----------------------------------------------------
    # Validation
    # -----------------------------------------------------

    assert len(
        efficient_frontier_local
    ) >= int(
        frontier_points * 0.80
    )

    assert np.allclose(
        efficient_frontier_weights_local.sum(
            axis=1
        ),
        1.0,
        atol=1e-6,
    )

    assert (
        efficient_frontier_weights_local
        .max()
        .max()
        <= maximum_stock_weight + 1e-5
    )

    return {
        "frontier":
            efficient_frontier_local,

        "frontier_weights":
            efficient_frontier_weights_local,

        "best_frontier_result":
            best_frontier_result,

        "best_frontier_weights":
            best_frontier_weights,

        "comparison_points":
            comparison_points_local,

        "minimum_feasible_return":
            minimum_feasible_return,

        "maximum_feasible_return":
            maximum_feasible_return,
    }


# =========================================================
# TEST ON THE CUSTOM SIX-STOCK PORTFOLIO
# =========================================================

custom_frontier_results = (
    build_arbitrary_efficient_frontier(
        expected_returns=(
            custom_optimisation_results[
                "expected_returns"
            ]
        ),
        covariance_matrix=(
            custom_optimisation_results[
                "covariance"
            ]
        ),
        comparison_weights=(
            custom_optimisation_results[
                "weights"
            ]
        ),
        minimum_stock_weight=0.00,
        maximum_stock_weight=0.30,
        risk_free_rate=RISK_FREE_RATE,
        frontier_points=60,
    )
)

custom_efficient_frontier = (
    custom_frontier_results[
        "frontier"
    ]
)

custom_best_frontier_result = (
    custom_frontier_results[
        "best_frontier_result"
    ]
)

custom_best_frontier_weights = (
    custom_frontier_results[
        "best_frontier_weights"
    ]
)

custom_frontier_figure = go.Figure()

custom_frontier_figure.add_trace(
    go.Scatter(
        x=(
            custom_efficient_frontier[
                "Volatility"
            ]
            * 100
        ),
        y=(
            custom_efficient_frontier[
                "Expected Return"
            ]
            * 100
        ),
        mode="lines+markers",
        name="Efficient Frontier",
        hovertemplate=(
            "Volatility: %{x:.2f}%"
            "<br>Expected return: %{y:.2f}%"
            "<extra></extra>"
        ),
    )
)

for portfolio_name, row in (
    custom_frontier_results[
        "comparison_points"
    ].iterrows()
):

    custom_frontier_figure.add_trace(
        go.Scatter(
            x=[
                row["Volatility"]
                * 100
            ],
            y=[
                row["Expected Return"]
                * 100
            ],
            mode="markers+text",
            name=portfolio_name,
            text=[
                portfolio_name
            ],
            textposition="top center",
        )
    )

custom_frontier_figure.update_layout(
    title=(
        "Custom Portfolio Constrained Efficient Frontier"
    ),
    xaxis_title=(
        "Model Annualised Volatility (%)"
    ),
    yaxis_title=(
        "Model Expected Return (%)"
    ),
    height=650,
)

custom_frontier_figure.show()


print("Generic efficient frontier completed.")
print("-" * 72)
print(
    "User portfolio holdings:",
    len(
        custom_best_frontier_weights
    ),
)
print(
    "Successful frontier portfolios:",
    len(
        custom_efficient_frontier
    ),
)
print(
    "Feasible expected-return range:",
    (
        f"{custom_frontier_results['minimum_feasible_return']:.2%}"
        " to "
        f"{custom_frontier_results['maximum_feasible_return']:.2%}"
    ),
)
print(
    "Best frontier expected return:",
    (
        f"{custom_best_frontier_result['Expected Return']:.2%}"
    ),
)
print(
    "Best frontier volatility:",
    (
        f"{custom_best_frontier_result['Volatility']:.2%}"
    ),
)
print(
    "Best frontier Sharpe:",
    (
        f"{custom_best_frontier_result['Sharpe Ratio']:.2f}"
    ),
)
print(
    "Best frontier active holdings:",
    int(
        custom_best_frontier_result[
            "Active Holdings"
        ]
    ),
)

print("\nBEST FRONTIER WEIGHTS")

display(
    custom_best_frontier_weights
    .sort_values(
        ascending=False
    )
    .to_frame("Weight")
    .style.format(
        {
            "Weight": "{:.2%}",
        }
    )
)
```
